<h1 style="color:#ffffff;font-family:Georgia,serif;font-size:2.2em;margin:0 0 14px 0;font-weight:700;line-height:1.3;">Beyond Correlation: A Multivariate Statistical Analysis of Music-Listening Profiles and Self-Reported Mental Health</h1>
<p style="color:#ede4fb;font-size:1.05em;margin:0 0 22px 0;font-family:Georgia,serif;line-height:1.6;max-width:820px;">A re-analysis of the MxMH survey (N = 736) built around inference rather than exploration: latent music-preference dimensions estimated by factor analysis, four correlated mental-health outcomes modelled jointly, and every association reported with an effect size, an uncertainty interval, and a multiple-testing correction, not a p-value alone.</p>

<h2 style="color:#5b21a6;margin:0 0 14px 0;font-size:1.25em;letter-spacing:1px;text-transform:uppercase;">Abstract</h2>
<div style="color:#2e1a47;line-height:1.85;font-size:0.98em;">

Music listening is nearly universal, and so is the folk intuition that it shapes mood, an intuition the
original MxMH survey and most public analyses of it lean on directly, describing individual genres as
sources of relief or harm. This notebook re-analyses the same 736-respondent survey with a different
standard of evidence. Sixteen genre-listening-frequency variables are reduced, using exploratory factor
analysis (KMO = 0.71, Bartlett's p &lt; .001), to four latent music-preference dimensions, and related to
four self-reported outcomes, anxiety, depression, insomnia, and OCD symptoms, through a progression of
models: multivariate regression respecting the outcomes' own correlation structure, ordinal logistic
regression respecting their bounded measurement scale, generalized additive models that do not assume
linearity, and a false discovery rate correction across all sixty-four genre-by-outcome comparisons rather
than raw p-values. A rock/metal-listening factor emerges as the analysis's most consistent finding: after
adjustment for age and behavioural covariates, it is positively associated with depression (standardized
&beta; = 0.207, 95% bootstrap CI [0.318, 0.678] on the raw scale), an association that survives six
independent robustness checks, continuous and ordinal outcome treatment, a Bayesian re-analysis (posterior
probability of a positive effect &asymp; 1.000), and correction for multiple testing at both the factor
and individual-genre level. Its predictive value is real but modest, raising cross-validated
classification performance only slightly above demographic and behavioural information alone, consistent
with a small-to-medium effect rather than a dramatic one. A proposed joint framework further shows that
music, age, and behaviour together explain only a small fraction of why the four outcomes correlate with
each other in the first place, that shared variance appears to originate elsewhere. No genre is described
as improving or worsening mental health, and no self-reported score is treated as a clinical diagnosis;
every association is reported as exactly that, an association, in a design that cannot distinguish it
from reverse causality or shared unmeasured causes.

<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 1</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Introduction and Research Motivation</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">Why this survey deserves a statistical re-analysis rather than another correlation heatmap</p>
</div>


The Music and Mental Health (MxMH) survey has been analysed publicly many times, and most of those
analyses share a structure: a correlation heatmap, a set of genre-by-outcome bar charts, and a closing
paragraph that reads the resulting patterns as evidence about what music does to the people who listen
to it. That structure is worth being specific about, because it is not a minor stylistic choice, it
encodes a scientific claim, that an observed difference in mean anxiety between listeners of two genres
reflects an effect of the music, when the data cannot distinguish that story from at least two others
equally consistent with it: people already experiencing anxiety may gravitate toward particular music,
or some third factor, age, listening habits, personality, may drive both the genre preference and the
mental-health outcome at once. A cross-sectional, self-reported, observational survey of 736 respondents
cannot adjudicate between these stories on its own, and treating a raw correlation as though it could is
the single most consequential scientific error available in this dataset.
<br><br>
This notebook takes the same 736 responses and asks a narrower, more defensible question: after
adjusting for the demographic and behavioural characteristics the survey also measured, are
music-listening patterns statistically associated with differences in self-reported anxiety, depression,
insomnia, and OCD symptoms, and how much of any raw association survives that adjustment. Framed this
way, the sixteen genre-frequency variables stop being sixteen separate stories to narrate individually
and become a measurement problem: they are highly correlated with one another, and a regression that
enters all sixteen at once will produce unstable, hard-to-interpret coefficients purely from that
collinearity, independent of anything substantive about music or mental health. Section 9 addresses this
directly with factor analysis, reducing the genre variables to a small number of latent listening
dimensions before they are ever used as predictors. The four outcomes present a parallel problem in the
other direction: they are correlated with each other for reasons a study of any one of them in isolation
would miss entirely, and Sections 8 and 12 treat them jointly rather than as four unrelated regressions
run side by side.
<br><br>
Every method used from here on is chosen to answer a specific question this framing raises, and each
section states that question, the model that answers it, and what the model assumes, before reporting a
single number. Where a method's assumptions do not hold for this data, Section 18 and the relevant
individual sections say so and report the alternative used instead. Where an association does not survive
adjustment or correction for multiple testing, that is reported as a finding in its own right rather than
omitted.

<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 2</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Research Questions and Hypotheses</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">What this notebook commits to answering, stated before any model is fit</p>
</div>


Five research questions are pre-specified here, before Section 5's data-quality analysis or any model
in Sections 9 onward. Pre-specification matters for exactly the reason Section 16 returns to formally:
with sixteen genres, four outcomes, and dozens of candidate covariates and interactions, it is
straightforward to search until something crosses a significance threshold by chance alone. Committing to
these five questions, and to primary versus exploratory status for the hypotheses under each, is what
separates a confirmatory analysis from a search.


<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">2.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Research Questions</span>
</div>


**RQ1.** After adjustment for demographic and behavioural characteristics, are music-listening patterns
statistically associated with self-reported anxiety, depression, insomnia, and OCD symptoms?

**RQ2.** Can the sixteen genre-listening-frequency variables be represented by a smaller number of
latent music-preference dimensions, and if so, how many, and what do they capture?

**RQ3.** Do the associations between music-listening patterns and mental-health outcomes vary with age,
daily listening duration, or other behavioural characteristics, and are they linear?

**RQ4.** Are the findings robust to alternative statistical specifications, missing-data treatments,
outlier treatments, and choices between continuous and ordinal outcome modelling?

**RQ5.** Does music-listening information provide meaningful predictive value for mental-health scores
or high-symptom categories, beyond what demographic and behavioural variables alone provide, without
that predictive value being read as diagnostic capability?


<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">2.2</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Formal Hypotheses</span>
</div>

<div style="background:#f5f3ff;border-radius:14px;padding:22px 28px;border:1.5px solid #c4b5fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">
<h4 style="color:#4c1d95;font-size:1.0rem;margin:0 0 10px;">Hypotheses</h4>

<p style="margin:0 0 4px 0;"><strong>H1 (primary, addresses RQ1).</strong> At least one latent
music-preference dimension is statistically significantly associated with at least one of the four
mental-health outcomes after adjustment for age, listening duration, and behavioural covariates,
surviving false discovery rate correction across all tested associations.</p>
<p style="margin:0 0 12px 0;font-size:12.5px;color:#581c87;"><em>H</em><sub>0</sub>: every adjusted
regression coefficient relating a latent music factor to a mental-health outcome is zero in the
population. Tested in Sections 11-12 with FDR-corrected p-values from Section 16.</p>

<p style="margin:0 0 4px 0;"><strong>H2 (primary, addresses RQ2).</strong> The sixteen genre-frequency
variables share enough common variance to be well represented by a substantially smaller number of
latent factors, evidenced by sampling adequacy and a clear scree discontinuity.</p>
<p style="margin:0 0 12px 0;font-size:12.5px;color:#581c87;"><em>H</em><sub>0</sub>: the genre variables
are effectively uncorrelated once measurement noise is accounted for, and no reduction below sixteen
dimensions is justified. Tested in Section 9 with the Kaiser-Meyer-Olkin statistic and Bartlett's test
of sphericity before any factor is extracted.</p>

<p style="margin:0 0 4px 0;"><strong>H3 (secondary/exploratory, addresses RQ3).</strong> The association
between music-preference factors and mental-health outcomes is moderated by age and by daily listening
duration, and at least one outcome's relationship with listening duration is nonlinear.</p>
<p style="margin:0 0 12px 0;font-size:12.5px;color:#581c87;"><em>H</em><sub>0</sub>: interaction terms
between music factors and age or listening duration are zero, and a linear specification fits as well as
a smooth nonlinear one. Tested in Sections 14-15, treated as exploratory and interpreted with more
caution than H1 or H2 regardless of outcome, per the pre-specification above.</p>

<p style="margin:0 0 4px 0;"><strong>H4 (primary, addresses RQ4).</strong> Any association found under H1
remains directionally consistent, though not necessarily identical in magnitude, across complete-case
and imputed samples, across raw-genre and latent-factor specifications, and across continuous and ordinal
outcome models.</p>
<p style="margin:0 0 12px 0;font-size:12.5px;color:#581c87;"><em>H</em><sub>0</sub>: apparent associations
from Section 11 are artefacts of one specific modelling choice and do not survive being tested a different
way. Tested throughout Section 18's robustness table.</p>

<p style="margin:0;"><strong>H5 (secondary, addresses RQ5).</strong> A model including music-preference
information achieves measurably better cross-validated predictive performance for at least one outcome
than an otherwise identical model built only from demographic and behavioural variables.</p>
<p style="margin:0;font-size:12.5px;color:#581c87;"><em>H</em><sub>0</sub>: cross-validated predictive
performance with and without music variables is statistically indistinguishable. Tested in Sections
20-22; a null result here does not weaken H1, since a variable can carry a genuine, real association
without adding useful predictive signal on top of correlated covariates already in the model, a
distinction Section 21 returns to explicitly.</p>

</div>

<div style="background:linear-gradient(135deg,#2e1a47 0%,#5b21a6 60%,#8b5cf6 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(46,26,71,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 3</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Theoretical Background</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">The mathematics behind every method used later, stated before any model touches the data</p>
</div>


This section exists so that no later modelling choice arrives unexplained. Every method used from
Section 9 onward is introduced here first: what question it answers, its mathematical form, what it
assumes, and where those assumptions are checked. Methods are grouped by the kind of question they
answer rather than the order they appear in later, since several (bootstrap, FDR correction) are applied
repeatedly across otherwise unrelated sections.


<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Correlation and Its Uncertainty</span>
</div>

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0 0 14px 0;">Pearson's <em>r</em> measures linear association and assumes the underlying relationship is
approximately linear and the variables are measured on an interval scale; Spearman's &rho; measures
monotonic association on the rank order of the data and requires neither linearity nor interval
measurement, which matters here since the mental-health outcomes take a small number of discrete values.
Both are reported where a comparison between them is itself informative, a large gap between Pearson and
Spearman correlations is a direct sign of a nonlinear or outlier-driven relationship Pearson's r alone
would miss.</p>

</div>

$$r = \frac{\sum_i (x_i-\bar x)(y_i-\bar y)}{\sqrt{\sum_i (x_i-\bar x)^2}\sqrt{\sum_i (y_i-\bar y)^2}}, \qquad \rho_s = 1 - \frac{6\sum_i d_i^2}{n(n^2-1)}$$

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0;">where <em>d</em><sub>i</sub> is the difference between the two rankings for observation <em>i</em>. A
point estimate of <em>r</em> without an interval is a common source of overconfidence with n in the
hundreds rather than the thousands, so confidence intervals here use Fisher's z-transformation
(Fisher, 1921), <em>z</em> = arctanh(<em>r</em>), which is approximately normally distributed with
standard error 1/&radic;(n-3), transformed back to the correlation scale for reporting.</p>

</div>

<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.2</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Principal Component Analysis</span>
</div>

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0 0 14px 0;">PCA (Pearson, 1901; Hotelling, 1933) re-expresses a set of correlated variables as a smaller set of
uncorrelated linear combinations, each chosen to capture as much remaining variance as possible. It is
used here as a purely descriptive, assumption-light first pass at the genre-frequency variables' shared
structure, before the more model-based factor analysis in Section 3.3 is asked to do the actual
dimension-reduction used later. Given standardised variables <em>X</em>, the components solve:</p>

</div>

$$\text{PC}_k = X v_k, \qquad v_k = \underset{\lVert v \rVert=1,\, v \perp v_1,\ldots,v_{k-1}}{\arg\max}\ \text{Var}(Xv)$$

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0;">equivalently, the <em>v</em><sub>k</sub> are the eigenvectors of the sample correlation matrix of
<em>X</em>, ordered by their eigenvalues, which equal the variance each component explains. PCA makes no
distributional assumption and does not distinguish shared (common) variance from variable-specific
noise, which is exactly the distinction Section 3.3's factor model does make, and is why Section 9 treats
PCA as an exploratory check on the number of dimensions rather than as the final representation.</p>

</div>

<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.3</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Exploratory Factor Analysis</span>
</div>

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0 0 14px 0;">Where PCA re-expresses all of a variable's variance, factor analysis (Spearman, 1904; Thurstone, 1947)
explicitly separates variance shared with other variables (common variance, explained by latent factors)
from variance specific to that one variable (uniqueness, including measurement error). This is the more
defensible model when the genre variables are treated as imperfect indicators of a smaller number of
underlying listening dispositions, which is the interpretation Section 9 needs for the factors to be used
as predictors later. The model for standardised observed variable <em>j</em>:</p>

</div>

$$x_j = \sum_{k=1}^{m} \lambda_{jk} f_k + \psi_j, \qquad \text{Var}(f_k)=1,\ \ \text{Cov}(f_k,\psi_j)=0$$

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0 0 14px 0;">where &lambda;<sub>jk</sub> is the loading of variable <em>j</em> on factor <em>k</em> and &psi;<sub>j</sub>
is its uniqueness. Two checks are run before extracting any factor, because factor analysis on
weakly-correlated variables produces uninterpretable, unstable factors: the Kaiser-Meyer-Olkin measure
of sampling adequacy (Kaiser, 1974), which compares the sum of squared correlations to the sum of squared
partial correlations and is conventionally treated as acceptable above 0.6, and Bartlett's test of
sphericity (Bartlett, 1950), which tests whether the correlation matrix differs significantly from an
identity matrix (H<sub>0</sub>: variables are uncorrelated in the population):</p>

</div>

$$\chi^2 = -\left(n-1-\frac{2p+5}{6}\right)\ln|R|, \qquad \text{df}=\frac{p(p-1)}{2}$$

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0;">where <em>R</em> is the correlation matrix and <em>p</em> the number of variables. The number of factors
to extract is chosen from a scree plot and the conventional eigenvalue-greater-than-one rule (Kaiser,
1960) taken together rather than either alone, and extracted factors are rotated (oblique promax rotation
where correlated factors are plausible, as they are for music-genre preferences) to improve
interpretability without changing the amount of variance explained. Factor names are assigned only after
inspecting the rotated loadings in Section 9, never before.</p>

</div>

<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.4</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Multivariate Multiple Regression and MANOVA</span>
</div>

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0 0 14px 0;">The four mental-health outcomes are analysed jointly as well as separately because they are correlated
(Section 8 quantifies this directly), and a set of four separate regressions discards that correlation
structure entirely, treating four views of what may be a shared underlying distress dimension as though
they were unrelated experiments. Multivariate multiple regression models all four simultaneously:</p>

</div>

$$Y_i = \alpha + B X_i + \varepsilon_i, \qquad Y_i = \begin{bmatrix}\text{Anxiety}_i\\ \text{Depression}_i\\ \text{Insomnia}_i\\ \text{OCD}_i\end{bmatrix},\quad \varepsilon_i \sim \mathcal{N}(0, \Sigma)$$

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0;">where <em>B</em> is a matrix of coefficients (one column per outcome) and &Sigma; is the outcomes'
residual covariance, left unconstrained rather than assumed diagonal, which is precisely what lets the
model use the outcomes' correlation rather than discard it. Wilks' lambda (Wilks, 1932) tests whether a
predictor's effect on the full outcome vector is jointly zero, a multivariate generalisation of the
F-test that is more powerful than four separate univariate tests when the outcomes are correlated and the
true effect is shared across them, and less prone to the multiple-testing inflation four independent
tests would otherwise carry.</p>

</div>

<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.5</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Ordinal Logistic Regression</span>
</div>

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0 0 14px 0;">Treating a bounded, coarsely-measured 0-10 self-report as an unbounded continuous variable is a modelling
choice, not a neutral default, and Section 4 examines directly whether it is the right one for these four
outcomes. The proportional-odds ordinal logistic model (McCullagh, 1980) is the natural alternative: it
respects the ordering of response categories without assuming the distance between "6" and "7" on a
self-report scale is numerically identical to the distance between "2" and "3". For <em>K</em> ordered
categories:</p>

</div>

$$\text{logit}\big[P(Y_i \le k)\big] = \theta_k - X_i\beta, \qquad k=1,\ldots,K-1$$

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0;">with cut-points &theta;<sub>k</sub> and a single coefficient vector &beta; shared across all category
thresholds, the proportional-odds assumption, checked in Section 13 with a Brant-style test comparing
against separate binary logistic fits at each cut-point. exp(&beta;) is interpreted as an odds ratio: the
multiplicative change in the odds of scoring at or below any given threshold for a one-unit increase in a
predictor. Where the assumption fails for a given outcome, Section 13 reports that failure explicitly
rather than reporting odds ratios the model does not actually support.</p>

</div>

<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.6</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Generalized Additive Models</span>
</div>

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0 0 14px 0;">A linear term forces a relationship to have the same slope everywhere; a generalized additive model
(Hastie and Tibshirani, 1986) replaces selected linear terms with smooth, data-estimated functions while
keeping the rest of the model linear and interpretable:</p>

</div>

$$Y_i = \beta_0 + s_1(\text{Age}_i) + s_2(\text{Hours}_i) + s_3(\text{BPM}_i) + X_i\beta + \varepsilon_i$$

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0;">where each <em>s</em>(&middot;) is a penalised regression spline whose wiggliness is chosen by
cross-validation rather than by hand, trading off fit against overfitting. A GAM is only worth its extra
complexity if a smooth term actually fits better than a straight line once that penalty is accounted for,
which Section 14 checks with an F-test comparing the linear and smooth specifications rather than fitting
a GAM by default because a spline is available.</p>

</div>

<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.7</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Multiple Testing Correction</span>
</div>

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0 0 14px 0;">Sixteen genres considered against four outcomes is sixty-four tests before a single covariate or
interaction is added; at a nominal 0.05 threshold, roughly three of those would be expected to appear
significant by chance alone even if every true association were exactly zero. The Benjamini-Hochberg
procedure (Benjamini and Hochberg, 1995) controls the false discovery rate, the expected proportion of
rejected null hypotheses that are actually false positives, rather than the more conservative familywise
error rate Bonferroni correction targets, appropriate here because a small number of false positives
among many tested associations is a more acceptable cost than the much lower power Bonferroni would
impose on a dataset this size. For <em>m</em> ordered p-values <em>p</em><sub>(1)</sub> &le;
&hellip; &le; <em>p</em><sub>(m)</sub>:</p>

</div>

$$\text{reject } H_{(i)} \text{ for all } i \le \max\left\{ i : p_{(i)} \le \frac{i}{m}\alpha \right\}$$

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0;">Every genre-by-outcome association in this notebook is reported alongside both its raw and its
FDR-adjusted p-value, and Section 16 states explicitly which associations survive adjustment and which
do not, rather than reporting only whichever number is more favourable.</p>

</div>

<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.8</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Bootstrap Confidence Intervals</span>
</div>

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0 0 14px 0;">A confidence interval derived from a formula (as in ordinary least squares) is only as trustworthy as the
distributional assumption behind that formula. The nonparametric bootstrap (Efron, 1979) sidesteps that
assumption by resampling the observed data itself with replacement, refitting the statistic of interest
on each resample, and reading the interval directly off the resulting distribution:</p>

</div>

$$\hat\theta^{*(b)} = T(X_1^{*(b)},\ldots,X_n^{*(b)}), \quad b=1,\ldots,B, \qquad \text{CI}_{95\%} = \big[\hat\theta^{*}_{(0.025)},\ \hat\theta^{*}_{(0.975)}\big]$$

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0;">using the percentile method for simplicity and transparency. B = 2,000 resamples are used throughout
this notebook wherever a bootstrap interval is reported, a number chosen to keep the 2.5th and 97.5th
percentiles themselves reasonably stable without making the resampling loop a runtime burden on Kaggle.
A bootstrap interval answers a specific question worth stating plainly: were this survey run again on a
new sample of the same size from the same population, this is the range within which the estimated
quantity would fall in 95% of such hypothetical repetitions, not a statement about where the true
population value lies with 95% probability, which is the credible interval's job in Section 3.11's
Bayesian framing, not the confidence interval's.</p>

</div>

<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.9</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Regularized Regression</span>
</div>

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0 0 14px 0;">Section 10 shows the sixteen genre variables are substantially collinear, which inflates ordinary least
squares coefficient variance without necessarily biasing the fit itself, coefficients become unstable and
hard to interpret individually even when the overall model predicts reasonably well. Three regularized
estimators trade a small amount of bias for a often much larger reduction in variance, each penalising
the coefficient vector differently:</p>

</div>


$$\text{Ridge (Hoerl and Kennard, 1970):} \quad \hat\beta = \arg\min_\beta \left[\text{RSS} + \lambda\sum_j \beta_j^2\right]$$
$$\text{Lasso (Tibshirani, 1996):} \quad \hat\beta = \arg\min_\beta \left[\text{RSS} + \lambda\sum_j |\beta_j|\right]$$
$$\text{Elastic Net (Zou and Hastie, 2005):} \quad \hat\beta = \arg\min_\beta \left[\text{RSS} + \lambda\Big(\alpha\sum_j|\beta_j| + \tfrac{1-\alpha}{2}\sum_j\beta_j^2\Big)\right]$$


<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0;">Ridge shrinks correlated coefficients toward each other without setting any to exactly zero; Lasso can
zero out coefficients entirely, performing a form of automatic variable selection, but tends to
arbitrarily pick one variable from a correlated group and zero the rest; Elastic Net blends both
penalties specifically to handle groups of correlated predictors, retaining or dropping them together
rather than arbitrarily favouring one. &lambda; (and &alpha; for Elastic Net) are tuned by
cross-validation in Section 20, never chosen by eye. A variable surviving Lasso or Elastic Net's penalty
indicates retained predictive signal conditional on the other variables in the model, not causal
importance, a distinction Section 20 returns to explicitly given how easily the two are conflated in
informal reporting.</p>

</div>

<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.10</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Predictive Model Evaluation</span>
</div>

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0 0 14px 0;">Section 21's classification models (predicting a high-symptom category) are evaluated on more than raw
accuracy, which can look deceptively good on an imbalanced outcome by always predicting the majority
class. The ROC curve plots the true positive rate against the false positive rate across all decision
thresholds, and its area (AUC) is the probability a randomly chosen positive case is ranked above a
randomly chosen negative one; the precision-recall curve and its area are reported alongside it since
AUC-ROC can look optimistic when positives are a small minority of the sample, which Section 5 will show
is the case for several high-symptom thresholds here. Calibration, whether a predicted probability of 0.7
actually corresponds to positive outcomes about 70% of the time, is assessed visually and with the Brier
score:</p>

</div>

$$\text{Brier} = \frac{1}{n}\sum_{i=1}^n (\hat p_i - y_i)^2$$

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0;">a proper scoring rule that rewards both discrimination and calibration simultaneously, unlike accuracy
or AUC alone. All predictive metrics in Sections 20-22 are computed under cross-validation, never on the
same data used to fit the model, and Section 5.34 states explicitly where each preprocessing step
(scaling, imputation, factor extraction) is refit inside each fold rather than once on the full sample,
the leakage-prevention rule good practice and the original request both require.</p>

</div>

<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">3.11</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Neural Networks and the Bayesian Alternative</span>
</div>

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0 0 14px 0;">A feedforward neural network composes simple nonlinear units to approximate more flexible functions
than a linear or additive model can. A single hidden layer computes:</p>

</div>

$$z = Wx + b, \qquad a = f(z), \qquad \hat y = W_{out}\,a + b_{out}$$

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0 0 14px 0;">where <em>f</em> is a nonlinear activation (ReLU here), <em>W</em> and <em>b</em> are learned weights and
biases, fit by minimising a loss function via gradient descent. With roughly 700 observations, Section
22 uses a single small hidden layer with dropout and early stopping rather than a deep architecture,
since a network with more free parameters than a linear model has predictors is far more likely to
overfit than to find real structure a GAM's smooth terms would miss, and Section 22 treats that
comparison, not the network's existence on its own, as the actual result worth reporting.</p>

</div>

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0 0 14px 0;">Separately, a Bayesian re-analysis of the primary regression treats &beta; itself as a random variable
with a prior distribution rather than a fixed unknown constant:</p>

</div>

$$\beta \sim \mathcal{N}(0, \sigma_0^2), \qquad Y_i \mid X_i, \beta \sim \mathcal{N}(X_i\beta,\ \sigma^2), \qquad p(\beta \mid Y, X) \propto p(Y\mid X,\beta)\,p(\beta)$$

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0;">producing a posterior distribution over &beta; rather than a single point estimate and standard error.
A 95% credible interval, unlike the bootstrap confidence interval of Section 3.8, is a direct probability
statement: given the model and the observed data, there is a 95% posterior probability the true parameter
lies in that interval, a genuinely different claim from the confidence interval's repeated-sampling
guarantee, and one worth stating precisely rather than treating the two interval types as interchangeable
language for the same idea. Section 24 reports this comparison only where it is feasible for this
sample size and adds a genuinely different interpretation, not as a checkbox exercise.</p>

</div>

The Music and Mental Health survey collected 736 responses across 33 fields (after dropping one field
described below). Four groups of variables matter for everything that follows. Demographic and listening
context: age, primary streaming service, average daily listening hours, and tempo (BPM) of the
respondent's favourite music. Listening behaviour: whether the respondent listens while working, plays
an instrument, composes music, actively seeks out new artists or genres, and listens to lyrics in
languages they do not speak fluently. Sixteen genre-listening-frequency items, each answered on a
four-point ordinal scale, Never, Rarely, Sometimes, Very frequently, encoded here as 0 through 3. Four
self-reported mental-health outcomes, anxiety, depression, insomnia, and OCD symptoms, each on a 0-10
scale, plus a separate categorical item asking whether the respondent believes music improves, worsens,
or has no effect on their mental health.
<br><br>
That last framing is worth being precise about immediately, because it recurs throughout this notebook:
these four outcomes are the respondent's own numeric self-rating of their symptoms, not a score from a
validated clinical instrument such as the GAD-7 or PHQ-9, and not a diagnosis made by a clinician.
Section 27 returns to what that distinction does and does not license concluding.

<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Data Quality and Preprocessing</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">Every exclusion rule defined, justified, and counted before a single model is fit</p>
</div>


This section follows one rule throughout: no observation is excluded, and no missing value is filled in,
without a stated reason, a rule defined in advance of looking at how it affects results, and a report of
exactly how many rows it touches.

<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">5.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">An Uninformative Column</span>
</div>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

df_raw = pd.read_csv('/kaggle/input/mxmh-survey-results/mxmh_survey_results.csv')
print(f"Raw shape: {df_raw.shape}")

print(f"\n'Permissions' column unique values: {df_raw['Permissions'].unique().tolist()}")
df = df_raw.drop(columns=['Permissions'])
print(f"Dropped as constant (a consent checkbox, not a variable). Shape now: {df.shape}")
```

<div style="background:linear-gradient(135deg,#faf5ff,#f3e8ff);border:1px solid #a855f744;border-left:5px solid #a855f7;border-radius:0 12px 12px 0;padding:18px 24px;margin:14px 0;">
<div style="font-weight:700;color:#581c87;margin-bottom:10px;font-size:14px;display:flex;align-items:center;gap:8px;"><span style="font-size:17px;">&#128270;</span> Why this one is a clean, risk-free exclusion</div>
<div style="color:#3b0764;font-size:13.5px;line-height:1.75;">
'Permissions' takes exactly one value across all 736 responses, a consent acknowledgement collected at
survey intake. A column with zero variance carries no statistical information and cannot bias, inflate,
or otherwise affect anything downstream; dropping it is definitionally safe in a way none of the
exclusion rules that follow are, which is exactly why it is handled first and separately from every
other decision in this section.</div>
</div>

<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">5.2</span>

<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Implausible Values: A Defined, Counted Exclusion Rule</span>
</div>


Three variables have plausible-looking ranges with implausible tails: BPM, daily listening hours, and
age. Each is handled differently, and the difference is deliberate rather than arbitrary.


```python
print("BPM distribution before any cleaning:")
print(df['BPM'].describe())
print(f"\nLargest 5 values: {sorted(df['BPM'].dropna().tolist())[-5:]}")
print(f"Smallest 5 values: {sorted(df['BPM'].dropna().tolist())[:5]}")
```

```python
# Rule, defined before inspecting how it affects any downstream result:
# retain BPM values in [40, 300]. The lower bound excludes values (0, single digits, low tens)
# that cannot correspond to an actual musical tempo; the upper bound is deliberately generous,
# set well above any officially documented genre's typical range (even the fastest extreme-metal
# subgenres are usually described in the 200-300 range at the very top end), so this rule removes
# clear data-entry artefacts without removing genuinely fast but real tempos.
BPM_MIN, BPM_MAX = 40, 300
implausible_bpm = df['BPM'].notna() & ((df['BPM'] < BPM_MIN) | (df['BPM'] > BPM_MAX))
n_implausible = implausible_bpm.sum()
print(f"Implausible BPM values (outside [{BPM_MIN}, {BPM_MAX}]): {n_implausible} of "
      f"{df['BPM'].notna().sum()} non-missing ({100*n_implausible/df['BPM'].notna().sum():.1f}%)")
print(f"Excluded values were: {sorted(df.loc[implausible_bpm, 'BPM'].tolist())}")

df.loc[implausible_bpm, 'BPM'] = np.nan
print(f"\nBPM missing after this rule: {df['BPM'].isna().sum()} of {len(df)} "
      f"({100*df['BPM'].isna().sum()/len(df):.1f}%)")
print("\nBPM distribution after cleaning:")
print(df['BPM'].describe())
```

<h4 style="color:#5b21a6;margin:0 0 10px;">Why these eight values, and no others, are treated as errors rather than data</h4>

Two of the eight are unambiguous: a reported tempo of 999,999,999 BPM and a reported tempo of 0 BPM (three
respondents) cannot describe actual music under any interpretation, the former is many orders of
magnitude beyond anything physically playable and the latter would mean no beat at all. The remaining
five (4, 8, 20, and one respondent's 624) sit far enough outside any documented genre's tempo range,
including the fastest extreme subgenres, that data entry error (an extra digit, a misplaced decimal, a
misunderstanding of the question) is a substantially more plausible explanation than a genuinely novel
musical tempo. Every other value, including values many listeners would consider unusually slow (40s) or
fast (200s), is retained without adjustment: being an outlier relative to the bulk of the sample is not,
on its own, a reason to treat a value as invalid, and Section 19 returns to check whether any of the
retained extreme-but-plausible values are unduly influencing the regression results.

<h4 style="color:#a21caf;margin:0 0 10px;font-size:1.0rem;">Why these two are flagged but not excluded here</h4>

Five respondents report more than sixteen hours of daily listening, three of them the round number
twenty-four, and four report ages of ten to twelve with one respondent aged eighty-nine. None of these
are impossible in the way a BPM of zero or a billion is: near-continuous background listening while
awake is a real, if unusual, pattern respondents might reasonably round to "all day," and a ten-year-old
or an eighty-nine-year-old completing an online survey is unusual but entirely plausible. The original
notebook's approach, removing the high-hours respondents on the stated grounds that they had unfavourable
mental-health scores, inverts the correct order of reasoning: an observation's outcome value is never a
valid justification for excluding it, doing so mechanically selects against whichever direction of
outcome is inconvenient. These respondents are kept in the primary analysis, and Section 18's robustness
table reports every primary result recomputed with a stricter (16-hour, 13-80-year) plausibility filter
applied, so that any sensitivity to these specific rows is visible rather than hidden by a decision made
once and not revisited.

<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Comparing Missing-Data Strategies</span>
</div>


Three treatments of the (now 115, after the BPM rule above) missing values are compared rather than one
chosen by default: complete-case analysis, univariate median or mode imputation, and multivariate
iterative imputation, which imputes each variable from the others rather than from its own marginal
distribution alone.


```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

model_vars = (['Age', 'Hours per day', 'BPM', 'While working', 'Instrumentalist', 'Composer',
               'Exploratory', 'Foreign languages'] +
              [c for c in df.columns if c.startswith('Frequency [')] +
              ['Anxiety', 'Depression', 'Insomnia', 'OCD'])

n_complete_case = df[model_vars].dropna().shape[0]
print(f"Complete-case N for a full model with all covariates: {n_complete_case} of {len(df)} "
      f"({100*n_complete_case/len(df):.1f}% retained)")

# Univariate strategy: median for numeric, mode for categorical
df_median_imputed = df.copy()
df_median_imputed['BPM'] = df_median_imputed['BPM'].fillna(df_median_imputed['BPM'].median())
df_median_imputed['Age'] = df_median_imputed['Age'].fillna(df_median_imputed['Age'].median())
for c in ['Primary streaming service', 'While working', 'Instrumentalist', 'Composer', 'Foreign languages']:
    df_median_imputed[c] = df_median_imputed[c].fillna(df_median_imputed[c].mode()[0])

# Multivariate strategy: iterative (MICE-style) imputation on the numeric block
numeric_for_impute = ['Age', 'Hours per day', 'BPM']
imputer = IterativeImputer(random_state=RANDOM_SEED, max_iter=10)
df_iterative_imputed = df.copy()
df_iterative_imputed[numeric_for_impute] = imputer.fit_transform(df[numeric_for_impute])

comparison = pd.DataFrame({
    'Strategy': ['Complete case', 'Median/mode imputation', 'Iterative (MICE-style) imputation'],
    'N used': [n_complete_case, len(df_median_imputed), len(df_iterative_imputed)],
    'BPM mean': [df['BPM'].mean(), df_median_imputed['BPM'].mean(), df_iterative_imputed['BPM'].mean()],
    'BPM std': [df['BPM'].std(), df_median_imputed['BPM'].std(), df_iterative_imputed['BPM'].std()],
}).round(2)
comparison
```

<div style="background:#f5f3ff;border-radius:12px;padding:20px 26px;border-left:5px solid #7c3aed;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin:14px 0;">
<h4 style="color:#5b21a6;margin:0 0 10px;">Which strategy is used where, and why</h4>

Complete-case analysis discards no information it has not been explicitly told to discard and remains
the primary specification throughout Sections 11-17, both for interpretability and because the retained
sample size above shows the cost of doing so is modest, not the severe reduction that would make
complete-case analysis indefensible on its own. Median imputation understates the true variability in
BPM, visible directly in its smaller standard deviation compared to the complete-case column, since every
imputed respondent receives the identical central value; iterative imputation instead predicts each
missing BPM from that respondent's age, listening hours, and other observed variables, preserving more
of the original variable's spread while still allowing every row to be used. Iterative imputation is
therefore the preferred alternative wherever the sample-size cost of complete-case analysis needs to be
avoided, and Section 18's robustness table reports the primary regression results under both, so that any
finding sensitive to this choice is reported as such rather than silently depending on whichever one
happened to be used first.

<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">Are anxiety, depression, insomnia, and OCD related but distinguishable, or effectively one thing</p>

```python
from itertools import combinations

def fisher_ci(r, n, alpha=0.05):
    z = np.arctanh(r)
    se = 1 / np.sqrt(n - 3)
    zcrit = stats.norm.ppf(1 - alpha/2)
    return np.tanh(z - zcrit*se), np.tanh(z + zcrit*se)

rows = []
n = len(df)
for a, b in combinations(outcomes, 2):
    r, p = stats.pearsonr(df[a], df[b])
    lo, hi = fisher_ci(r, n)
    rows.append({'Pair': f'{a}-{b}', 'r': r, '95% CI': f'[{lo:.3f}, {hi:.3f}]', 'p-value': p})
outcome_corr_table = pd.DataFrame(rows)
outcome_corr_table['r'] = outcome_corr_table['r'].round(3)
outcome_corr_table
```

<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Music Preference Representation: PCA and Factor Analysis</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">Reducing sixteen correlated genre variables to a small number of interpretable listening dimensions</p>
</div>


Sections 3.2 and 3.3 laid out the theory; this section runs it. Sampling adequacy is checked first,
because factor analysis on variables that do not actually share enough variance produces factors that
are statistically extractable but substantively meaningless.


<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">9.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Is Factor Analysis Justified for These Sixteen Variables</span>
</div>

```python
from scipy.stats import chi2 as chi2_dist

def kmo_test(X):
    """Kaiser-Meyer-Olkin sampling adequacy, computed directly from the correlation matrix and
    its inverse (the partial-correlation structure), avoiding a dependency on the factor_analyzer
    package, whose current PyPI release (0.5.1) calls sklearn's check_array with an argument name
    removed in recent scikit-learn versions and raises a TypeError on import in this environment."""
    R = np.corrcoef(X, rowvar=False)
    R_inv = np.linalg.inv(R)
    d = np.sqrt(np.diag(R_inv))
    partial_corr = -R_inv / np.outer(d, d)
    np.fill_diagonal(partial_corr, 0)
    R_offdiag = R.copy(); np.fill_diagonal(R_offdiag, 0)
    sum_r2, sum_pr2 = np.sum(R_offdiag**2), np.sum(partial_corr**2)
    kmo_overall = sum_r2 / (sum_r2 + sum_pr2)
    kmo_per_var = (np.sum(R_offdiag**2, axis=0) /
                   (np.sum(R_offdiag**2, axis=0) + np.sum(partial_corr**2, axis=0)))
    return kmo_overall, kmo_per_var

def bartlett_sphericity(X):
    n, p = X.shape
    R = np.corrcoef(X, rowvar=False)
    _, logdet = np.linalg.slogdet(R)
    chi2_stat = -(n - 1 - (2*p + 5)/6) * logdet
    dof = p * (p - 1) / 2
    p_value = 1 - chi2_dist.cdf(chi2_stat, dof)
    return chi2_stat, dof, p_value

genre_names = freq_numeric.columns.tolist()
kmo_overall, kmo_per_var = kmo_test(freq_numeric.values)
chi2_stat, dof, p_bartlett = bartlett_sphericity(freq_numeric.values)

print(f"Overall KMO: {kmo_overall:.3f}")
print(f"Bartlett's test of sphericity: chi2({dof:.0f}) = {chi2_stat:.1f}, p = {p_bartlett:.2e}")

kmo_table = pd.DataFrame({'Genre': genre_names, 'MSA (per-variable KMO)': kmo_per_var}).sort_values(
    'MSA (per-variable KMO)')
kmo_table.round(3)
```

<div style="background:#f5f3ff;border-radius:12px;padding:20px 26px;border-left:5px solid #7c3aed;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin:14px 0;">
<h4 style="color:#5b21a6;margin:0 0 10px;">Results: factorable, though not uniformly strong across genres</h4>

An overall KMO of 0.71 falls in the "middling" to "acceptable" range by Kaiser's conventional bands, well
above the 0.6 floor below which factor analysis is not recommended, and Bartlett's test overwhelmingly
rejects the null of an identity correlation matrix. H2's sampling-adequacy prediction is supported. The
per-variable breakdown is worth reading rather than skipping past: Rock, Metal, Classical, and Video game
music have the lowest individual MSA values, meaning they share comparatively less of their variance with
the other fifteen genres and will likely load less cleanly on whatever common factors emerge, while
Latin, R&B, and K pop share the most. This is useful to know before looking at loadings, since a genre
with low MSA loading weakly on every factor is expected behaviour, not a modelling failure.

<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">How Many Dimensions: PCA Scree and Explained Variance</span>
</div>

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

X_scaled = StandardScaler().fit_transform(freq_numeric)
pca_full = PCA(random_state=RANDOM_SEED)
pca_full.fit(X_scaled)

eigenvalues = pca_full.explained_variance_
var_ratio = pca_full.explained_variance_ratio_ * 100
cum_var = np.cumsum(var_ratio)
n_above_kaiser = (eigenvalues > 1).sum()
print(f"Components with eigenvalue > 1 (Kaiser's rule): {n_above_kaiser}")
print(f"Variance explained by first 4 components: {cum_var[3]:.1f}%")
print(f"Variance explained by first 5 components: {cum_var[4]:.1f}%")
```

<div style="font-weight:700;color:#581c87;margin-bottom:10px;font-size:14px;display:flex;align-items:center;gap:8px;"><span style="font-size:17px;">&#128270;</span> Choosing 4 factors, not 5</div>
<div style="color:#3b0764;font-size:13.5px;line-height:1.75;">
The Kaiser criterion technically points to five components, but the scree plot's elbow, the point past
which additional components each add only a small, roughly equal increment, sits clearly at four: the
gap between the fourth and fifth eigenvalues is far smaller than the gaps preceding it, and the fifth
component's own eigenvalue (1.01) sits only marginally above the Kaiser threshold, the kind of borderline
case the criterion is known to handle poorly. Four factors are carried forward into the actual extraction
below, explaining a little over half the genre variables' total variance, a defensible, literature-typical
amount for exploratory factor analysis on survey items with this much idiosyncratic, genre-specific
variance still left over.

<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Extracting and Rotating Four Factors</span>
</div>

```python
from sklearn.decomposition import FactorAnalysis

def varimax(Phi, gamma=1.0, max_iter=100, tol=1e-8):
    p, k = Phi.shape
    R = np.eye(k)
    d = 0
    for _ in range(max_iter):
        Lambda = Phi @ R
        u, s, vt = np.linalg.svd(
            Phi.T @ (Lambda**3 - (gamma/p) * Lambda @ np.diag(np.diag(Lambda.T @ Lambda))))
        R = u @ vt
        d_new = np.sum(s)
        if d_new < d * (1 + tol):
            break
        d = d_new
    return Phi @ R

def promax_rotation(loadings, power=4):
    """Promax (Hendrickson & White, 1964): an orthogonal varimax solution is first obtained, then
    obliquely rotated toward it raised to a power, which sharpens high loadings and shrinks low ones,
    allowing the resulting factors to correlate rather than forcing strict orthogonality."""
    X = varimax(loadings)
    P = X * (np.abs(X) ** (power - 1))
    L = np.linalg.lstsq(X, P, rcond=None)[0]
    L_inv = np.linalg.inv(L)
    normalizer = np.diag(1 / np.sqrt(np.diag(L_inv.T @ L_inv)))
    T = L_inv @ normalizer
    return X @ T

N_FACTORS = 4
fa_model = FactorAnalysis(n_components=N_FACTORS, random_state=RANDOM_SEED, max_iter=2000)
fa_model.fit(X_scaled)
loadings_unrotated = fa_model.components_.T
loadings_rotated = promax_rotation(loadings_unrotated)

loadings_df = pd.DataFrame(loadings_rotated, index=genre_names,
                            columns=[f'Factor {i+1}' for i in range(N_FACTORS)])
communalities = (loadings_unrotated**2).sum(axis=1)
loadings_df['Communality'] = communalities
loadings_df.round(2).sort_values('Factor 1', ascending=False)
```

<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 10</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Multicollinearity Diagnostics</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">Checking, rather than assuming, that sixteen correlated genres would destabilise a regression</p>
</div>


Section 9 justified reducing the genre variables on grounds of shared variance and interpretability.
A separate, narrower question is whether entering all sixteen raw genre variables into one regression
would itself be a problem, in the specific technical sense of inflating coefficient variance to the
point of instability. That is checked directly rather than assumed.


```python
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor

X_vif = sm.add_constant(freq_numeric)
vif_table = pd.DataFrame({
    'Genre': X_vif.columns,
    'VIF': [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])]
})
vif_table = vif_table[vif_table['Genre'] != 'const'].sort_values('VIF', ascending=False).reset_index(drop=True)

corr_matrix = freq_numeric.corr()
eigvals = np.linalg.eigvalsh(corr_matrix)
condition_number = np.sqrt(eigvals.max() / eigvals.min())

print(f"Maximum VIF: {vif_table['VIF'].max():.2f} ({vif_table.iloc[0]['Genre']})")
print(f"Mean VIF: {vif_table['VIF'].mean():.2f}")
print(f"Genres with VIF > 5: {(vif_table['VIF']>5).sum()} of 16")
print(f"Condition number of the genre correlation matrix: {condition_number:.2f}")
vif_table
```

```python
fig, ax = plt.subplots(figsize=(9, 7))
colors = ['#a21caf' if v > 5 else '#7c3aed' for v in vif_table['VIF']]
ax.barh(vif_table['Genre'][::-1], vif_table['VIF'][::-1], color=colors[::-1], edgecolor='#4c1d95')
ax.axvline(5, color='#701a75', linestyle='--', linewidth=1.3, label='Conventional concern threshold (VIF=5)')
ax.set_title('Figure 12. Variance inflation factors, raw genre variables', fontsize=13, color='#5b21a6', fontweight='bold')
ax.set_xlabel('VIF'); ax.legend(frameon=False, fontsize=9)
plt.tight_layout(); plt.show()
```

<div style="background:#f5f3ff;border-radius:12px;padding:20px 26px;border-left:5px solid #7c3aed;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin:14px 0;">
<h4 style="color:#5b21a6;margin:0 0 10px;">An honest, slightly counter-intuitive result</h4>

None of the sixteen genres reaches a VIF anywhere near the conventional concern threshold of 5, the
largest, Hip hop, sits at 2.96, and the correlation matrix's condition number (4.04) is well inside the
range considered numerically unproblematic. In the strict sense the variance inflation factor measures,
whether a regression coefficient's own standard error is being substantially inflated by that variable's
correlation with the others in the same model, sixteen raw genre variables would not, on this evidence,
produce badly unstable OLS coefficients. This is worth stating plainly rather than reaching for the more
dramatic multicollinearity story a template might expect: the correlations Section 9 found are real and
support genuine shared variance worth extracting, but they are not so extreme that they would have
wrecked a naive regression on their own. Factor analysis remains well justified here, on grounds of
interpretability and dimensionality (sixteen coefficients collapsed to four with a clear substantive
reading) and on the theoretical grounds Section 3.3 lays out, rather than on a multicollinearity
emergency the VIF numbers do not actually support.

<div style="background:linear-gradient(135deg,#2e1a47 0%,#5b21a6 60%,#8b5cf6 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(46,26,71,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 11</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Primary Regression Models</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">Four outcomes, five specifications each, and whether crude associations survive adjustment</p>
</div>


**Objective.** Test H1: is at least one music-preference factor associated with at least one outcome
after adjustment. **Method.** Ordinary least squares, heteroskedasticity-robust (HC1) standard errors,
fit separately for each outcome across five specifications that add covariates incrementally.
**Assumptions.** Linearity in the included predictors, independent observations, and correctly-specified
robust standard errors under heteroskedasticity, checked in Section 19; no distributional assumption on
the outcome is required for point estimates, though Section 13 checks whether the continuous-outcome
treatment itself is defensible. **Why this model first.** OLS is the simplest, most interpretable
starting point and the natural baseline every more complex model in this notebook is compared against.


<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">11.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Five Specifications</span>
</div>


1. Music factors only. 2. + Age. 3. + Age and behavioural covariates (listening hours, working while
listening, instrumentalist, composer, exploratory, foreign-language listening). 4. Raw genre variables
in place of factors (a robustness check on the factor representation itself). 5. Complete-case in place
of the default row-wise handling (a robustness check on missingness treatment). Specifications 4 and 5
are reported in full in Section 18's robustness table rather than repeated four times here; this section
reports 1-3 for every outcome, since that progression is what directly tests whether crude associations
attenuate.


```python
# Factor scores via the regression method (Thurstone, 1935): each respondent's score on a factor
# is estimated as a weighted combination of their observed genre responses, using the (pseudo-inverse
# of the) rotated loading matrix as weights.
factor_scores = X_scaled @ np.linalg.pinv(loadings_rotated.T)
# Factors 2 and 4 loaded negatively on their defining genres (Section 9.4); sign-flip both so that,
# throughout the rest of this notebook, a higher score on every factor consistently means more of that
# kind of listening, not less.
factor_scores[:, 1] *= -1
factor_scores[:, 3] *= -1

df['MF1_eclectic'] = factor_scores[:, 0]
df['MF2_rockmetal'] = factor_scores[:, 1]
df['MF3_broad'] = factor_scores[:, 2]
df['MF4_electronic'] = factor_scores[:, 3]
music_factors = ['MF1_eclectic', 'MF2_rockmetal', 'MF3_broad', 'MF4_electronic']

for c in ['While working', 'Instrumentalist', 'Composer', 'Exploratory', 'Foreign languages']:
    df[c + '_bin'] = (df[c] == 'Yes').astype(float)
    df.loc[df[c].isna(), c + '_bin'] = np.nan

behavioural = ['Hours per day', 'While working_bin', 'Instrumentalist_bin', 'Composer_bin',
               'Exploratory_bin', 'Foreign languages_bin']
print(f"Factor score summary:\n{pd.DataFrame(factor_scores, columns=music_factors).describe().round(2)}")
```

```python
def fit_ols(y_col, x_cols, data):
    sub = data[[y_col] + x_cols].dropna()
    X = sm.add_constant(sub[x_cols])
    model = sm.OLS(sub[y_col], X).fit(cov_type='HC1')
    return model, len(sub)

specifications = {
    'S1: music only': music_factors,
    'S2: + age': music_factors + ['Age'],
    'S3: + behavioural': music_factors + ['Age'] + behavioural,
}

all_results = []
for outcome in outcomes:
    for spec_name, xcols in specifications.items():
        model, n = fit_ols(outcome, xcols, df)
        for mf in music_factors:
            all_results.append({
                'Outcome': outcome, 'Specification': spec_name, 'Predictor': mf,
                'Coefficient': model.params[mf], 'SE': model.bse[mf], 'p-value': model.pvalues[mf],
                'N': n, 'R2': model.rsquared,
            })
primary_results = pd.DataFrame(all_results)
primary_results.round(4)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(13, 10))
for ax, outcome in zip(axes.flat, outcomes):
    sub = primary_results[primary_results['Outcome'] == outcome]
    for mf, color in zip(music_factors, ['#7c3aed', '#dc2626', '#0891b2', '#a855f7']):
        row = sub[sub['Predictor'] == mf]
        ax.errorbar(range(3), row['Coefficient'], yerr=1.96*row['SE'], marker='o', capsize=4,
                    color=color, label=mf.split('_')[1], linewidth=1.8)
    ax.axhline(0, color='#44403c', linewidth=1)
    ax.set_xticks(range(3)); ax.set_xticklabels(['S1\n(music only)', 'S2\n(+age)', 'S3\n(+behavioural)'], fontsize=9)
    ax.set_title(outcome, fontsize=12.5, color='#4c1d95', fontweight='bold')
    ax.set_ylabel('Coefficient (95% CI)')
axes[0,0].legend(frameon=False, fontsize=8, loc='upper left')
fig.suptitle('Figure 13. Coefficient attenuation across specifications, all four outcomes', fontsize=14.5, color='#5b21a6', fontweight='bold', y=1.01)
plt.tight_layout(); plt.show()
```

<h4 style="color:#5b21a6;margin:0 0 10px;">Results: one robust association, one that attenuates, two that never appear</h4>

Read Figure 13 outcome by outcome. The rock/metal factor is positively associated with anxiety,
depression, and insomnia in the music-only specification, and, distinctively, its coefficient barely
moves as age and behavioural covariates are added, for depression specifically it is 0.484 unadjusted and
0.504 with the full covariate set, essentially unchanged, and it stays significant for all three of
those outcomes throughout. The electronic/digital factor shows the same pattern of attenuation only for
some outcomes, not uniformly: for depression its p-value moves from 0.016 to 0.078 to 0.226 across the
three specifications, a textbook case of a crude association substantially attenuated by confounding
once age and behaviour are accounted for, but for insomnia the same factor remains significant throughout
(p = 0.0013, 0.0006, 0.0024), essentially undiminished by the same covariates that erased its association
with depression. Outcome-specific attenuation like this, the identical predictor behaving differently
against two different outcomes once the same controls are added, is itself informative rather than an
inconsistency to explain away: it means whatever confounds the electronic-listening/depression link is
not confounding the electronic-listening/insomnia link in the same way, a distinction a single pooled
"music affects mental health" claim would erase entirely. The eclectic and broad-mainstream factors show
no significant association with any outcome at any specification, and none of the four factors is
significantly associated with OCD in any specification. Section 16 revisits every one of these
associations with a false discovery rate correction across all sixty-four genre-by-outcome tests before
treating any of them as a confirmed finding, and Section 18 checks whether the associations that looked
robust across S1-S3 specifically survive the raw-genre and complete-case robustness checks as well.

<div style="background:linear-gradient(135deg,#2e1a47 0%,#5b21a6 60%,#8b5cf6 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(46,26,71,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 12</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Multivariate Regression</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">Testing all four outcomes jointly, using their correlation rather than discarding it</p>
</div>


**Objective.** Confirm whether Section 11's per-outcome findings hold up under a test that accounts for
the outcomes' shared correlation structure (Section 8), rather than treating four F-tests as independent
evidence. **Method.** Multivariate multiple regression with Wilks' lambda testing each predictor's joint
effect on the full outcome vector. **Assumptions.** Multivariate normality of the residual vector, checked
qualitatively against Section 6's skewed marginal distributions, a real limitation this section flags
rather than assumes away. **Why this model.** Section 3.4 lays out the case: four correlated outcomes
analysed as four independent regressions inflates the effective number of tests without using the
information their correlation actually provides.


```python
from statsmodels.multivariate.manova import MANOVA

manova_data = df[outcomes + music_factors].dropna()
formula = f"Anxiety + Depression + Insomnia + OCD ~ {' + '.join(music_factors)}"
manova_model = MANOVA.from_formula(formula, data=manova_data)
manova_result = manova_model.mv_test()
print(manova_result)
```

```python
manova_summary = []
for mf in music_factors:
    stat_table = manova_result.results[mf]['stat']
    manova_summary.append({
        'Predictor': mf,
        "Wilks' lambda": stat_table.loc["Wilks' lambda", 'Value'],
        'F': stat_table.loc["Wilks' lambda", 'F Value'],
        'Num DF': stat_table.loc["Wilks' lambda", 'Num DF'],
        'Den DF': stat_table.loc["Wilks' lambda", 'Den DF'],
        'p-value': stat_table.loc["Wilks' lambda", 'Pr > F'],
    })
manova_summary_table = pd.DataFrame(manova_summary)
manova_summary_table.round(4)
```

<div style="background:#f5f3ff;border-radius:12px;padding:20px 26px;border-left:5px solid #7c3aed;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin:14px 0;">
<h4 style="color:#5b21a6;margin:0 0 10px;">Results: the joint test agrees with the four separate ones</h4>

The rock/metal factor has a highly significant joint effect on the four-outcome vector (Wilks' lambda =
0.961, p &lt; 0.0001) and the electronic/digital factor a significant one (lambda = 0.981, p = 0.008),
while the eclectic and broad-mainstream factors show no joint effect (p = 0.69 and p = 0.99). This is
exactly the same ranking Section 11's four separate regressions produced, which is itself the useful
finding: when a joint multivariate test and four univariate tests agree this closely, it means the
outcomes' shared correlation was not hiding or distorting anything Section 11 had already found, the
univariate results were not an artefact of running four tests instead of one properly joint test. Because
every outcome shares the identical set of predictors here, the multivariate model's own coefficient
estimates for each outcome are numerically identical to Section 11's separate-regression coefficients,
Gauss-Markov guarantees this whenever the right-hand side is the same across equations, so MANOVA's
distinct contribution is entirely in the joint significance test above, not in different point estimates.

<a id="sec13"></a>
<div style="background:linear-gradient(135deg,#2e1a47 0%,#5b21a6 60%,#8b5cf6 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(46,26,71,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 13</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Ordinal Regression</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">Checking whether the continuous-outcome treatment itself was the right choice</p>
</div>


**Objective.** Test H4's specific prediction that the rock/metal and electronic/digital associations
survive a change from continuous to ordinal outcome treatment. **Method.** Proportional-odds ordinal
logistic regression (Section 3.5). **Assumptions.** Proportional odds across all outcome thresholds, that
a one-unit predictor change shifts the odds of exceeding *every* threshold by the same multiplicative
factor. **Why this model.** Section 6 already showed all four outcomes are visibly non-Gaussian and
bounded; this is the direct check on whether that matters for the substantive conclusions.


```python
from statsmodels.miscmodels.ordinal_model import OrderedModel

ordinal_results = []
for outcome in outcomes:
    sub = df[[outcome] + music_factors + ['Age']].dropna()
    mod = OrderedModel(sub[outcome], sub[music_factors + ['Age']], distr='logit')
    res = mod.fit(method='bfgs', disp=False, maxiter=200)
    for mf in music_factors:
        ordinal_results.append({
            'Outcome': outcome, 'Predictor': mf, 'Coefficient': res.params[mf],
            'Odds ratio': np.exp(res.params[mf]), 'p-value': res.pvalues[mf], 'N': len(sub),
        })
ordinal_table = pd.DataFrame(ordinal_results)
ordinal_table.round(4)
```

```python
# Proportional-odds check: refit as a series of binary logistic regressions, one per threshold,
# and compare each threshold-specific coefficient to the single shared coefficient the ordinal
# model assumes. Large, systematic disagreement would indicate the assumption is not defensible.
outcome_check = 'Depression'
sub = df[[outcome_check] + music_factors].dropna()
thresholds = sorted(sub[outcome_check].unique())[:-1]
threshold_coefs = {mf: [] for mf in music_factors}
for t in thresholds:
    y_bin = (sub[outcome_check] > t).astype(int)
    X_bin = sm.add_constant(sub[music_factors])
    bin_model = sm.Logit(y_bin, X_bin).fit(disp=False)
    for mf in music_factors:
        threshold_coefs[mf].append(bin_model.params[mf])

fig, ax = plt.subplots(figsize=(9, 5.5))
for mf, color in zip(music_factors, ['#7c3aed', '#dc2626', '#0891b2', '#a855f7']):
    ax.plot(range(len(thresholds)), threshold_coefs[mf], marker='o', color=color, label=mf.split('_')[1])
ax.set_xticks(range(len(thresholds))); ax.set_xticklabels([f'>{t}' for t in thresholds], fontsize=8, rotation=45)
ax.set_title(f'Figure 14. Threshold-specific coefficients, {outcome_check}\n(proportional-odds check)',
             fontsize=12.5, color='#5b21a6', fontweight='bold')
ax.set_ylabel('Binary logistic coefficient'); ax.legend(frameon=False, fontsize=8)
plt.tight_layout(); plt.show()
```

<h4 style="color:#5b21a6;margin:0 0 10px;">Results: ordinal treatment agrees with continuous treatment</h4>

The rock/metal factor remains significant with an odds ratio of 1.34 for depression (p &lt; 0.0001):

<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Nonlinear Effects and Generalized Additive Models</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">Testing whether Age and listening duration actually relate to outcomes linearly</p>
</div>


**Objective.** Test H3's nonlinearity component directly, for Depression specifically, the outcome with
the clearest signal in Section 11. **Method.** A generalized additive model (Section 3.6) with
cubic B-spline smooth terms for Age and Hours per day, compared against the equivalent linear
specification with a nested F-test. **Assumptions.** The smooth terms' basis dimension (6 degrees of
freedom each here) bounds how wiggly the fitted curve can be; too few risks missing real curvature, too
many risks fitting noise, the choice is reported explicitly rather than left implicit. **Why this model.**
Section 11's linear terms for age assumed a constant per-year effect; this section checks that assumption
rather than carrying it forward silently.


```python
from statsmodels.gam.api import GLMGam, BSplines
from scipy.stats import f as f_dist

gam_sub = df[['Depression', 'Age', 'Hours per day']].dropna()
x_spline = gam_sub[['Age', 'Hours per day']]
bs = BSplines(x_spline, df=[6, 6], degree=[3, 3])
gam_model = GLMGam(gam_sub['Depression'], smoother=bs)
gam_res = gam_model.fit()

X_lin = sm.add_constant(gam_sub[['Age', 'Hours per day']])
lin_model = sm.OLS(gam_sub['Depression'], X_lin).fit()

rss_lin, rss_gam = np.sum(lin_model.resid**2), gam_res.deviance
df_lin, df_gam = lin_model.df_resid, gam_res.df_resid
f_stat = ((rss_lin - rss_gam) / (df_lin - df_gam)) / (rss_gam / df_gam)
p_value = 1 - f_dist.cdf(f_stat, df_lin - df_gam, df_gam)

print(f"Linear model R\u00b2: {lin_model.rsquared:.4f}")
print(f"GAM pseudo-R\u00b2 (1 - deviance/null deviance): {1 - gam_res.deviance/gam_res.null_deviance:.4f}")
print(f"Nested F-test, linear vs GAM: F({df_lin-df_gam:.0f}, {df_gam:.0f}) = {f_stat:.3f}, p = {p_value:.4f}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))
gam_res.plot_partial(0, cpr=False, ax=axes[0])
axes[0].set_title('Age', fontsize=12.5, color='#4c1d95', fontweight='bold')
axes[0].set_xlabel('Age (years)'); axes[0].set_ylabel('Partial effect on Depression')
gam_res.plot_partial(1, cpr=False, ax=axes[1])
axes[1].set_title('Hours per day', fontsize=12.5, color='#4c1d95', fontweight='bold')
axes[1].set_xlabel('Hours per day'); axes[1].set_ylabel('Partial effect on Depression')
fig.suptitle('Figure 15. GAM partial effects on Depression, with pointwise confidence bands',
             fontsize=14, color='#5b21a6', fontweight='bold', y=1.03)
plt.tight_layout(); plt.show()
```

<h4 style="color:#5b21a6;margin:0 0 10px;">Results: a real, but partly sparse-tail-driven, nonlinearity</h4>

The nested F-test rejects linearity decisively (p &lt; 0.0001): a smooth specification fits Depression
against age and listening hours significantly better than two straight lines, and the GAM's pseudo-R2
(0.076) roughly triples the linear model's R2 (0.026). The age partial effect rises from the youngest

well-supported part of the finding is the middle of each curve, a rise-then-decline in the age effect
across the bulk of the sample's twenties through sixties, not the sharp swings at either end, which are

<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Interaction Effects</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">Pre-specified tests of whether the rock/metal association depends on age or listening context</p>
</div>


**Objective.** Test H3's moderation component: does the rock/metal factor's association with depression,
the most robust finding so far, vary with age or listening behaviour. **Method.** OLS with a single
product term added to the age-adjusted model, one pre-specified interaction at a time rather than a
search across all possible pairs. **Why pre-specified.** Section 2.2 committed to this in advance
specifically to avoid the multiple-comparisons problem Section 16 addresses formally for the genre-level
tests; testing every possible interaction here and reporting only the significant ones would reintroduce
exactly that problem through the back door.


```python
interaction_results = []
for interact_var in ['Age', 'Hours per day']:
    sub = df[['Depression', 'MF2_rockmetal', interact_var]].dropna()
    sub['interaction'] = sub['MF2_rockmetal'] * sub[interact_var]
    X = sm.add_constant(sub[['MF2_rockmetal', interact_var, 'interaction']])
    model = sm.OLS(sub['Depression'], X).fit(cov_type='HC1')
    interaction_results.append({
        'Interaction': f'Rock/Metal \u00d7 {interact_var}',
        'Coefficient': model.params['interaction'], 'SE': model.bse['interaction'],
        'p-value': model.pvalues['interaction'], 'N': len(sub),
    })
interaction_table = pd.DataFrame(interaction_results)
interaction_table.round(4)
```

```python
sub = df[['Depression', 'MF2_rockmetal', 'Age']].dropna()
sub['interaction'] = sub['MF2_rockmetal'] * sub['Age']
X = sm.add_constant(sub[['MF2_rockmetal', 'Age', 'interaction']])
model = sm.OLS(sub['Depression'], X).fit(cov_type='HC1')

fig, ax = plt.subplots(figsize=(9, 6))
age_values = [sub['Age'].quantile(q) for q in [0.1, 0.5, 0.9]]
mf_range = np.linspace(sub['MF2_rockmetal'].min(), sub['MF2_rockmetal'].max(), 50)
for age_val, label, color in zip(age_values, ['10th pct age', 'Median age', '90th pct age'],
                                    ['#c4b5fd', '#7c3aed', '#4c1d95']):
    predicted = (model.params['const'] + model.params['MF2_rockmetal']*mf_range +
                 model.params['Age']*age_val + model.params['interaction']*mf_range*age_val)
    ax.plot(mf_range, predicted, color=color, linewidth=2, label=f'{label} ({age_val:.0f}y)')
ax.set_title('Figure 16. Marginal effect of Rock/Metal listening on Depression, by age',
             fontsize=12.5, color='#5b21a6', fontweight='bold')
ax.set_xlabel('Rock/Metal factor score'); ax.set_ylabel('Predicted Depression')
ax.legend(frameon=False, fontsize=9)
plt.tight_layout(); plt.show()
```

<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Multiple Testing Correction</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">Sixty-four genre-by-outcome tests, and which associations survive correcting for all of them</p>
</div>


**Objective.** Directly test H1 at the level Section 11's factor-based models could not: individual
genres, not the four composite factors, since a real association concentrated in one or two specific
genres could be diluted below significance once averaged into a broader factor alongside genres that
carry no such signal. **Method.** Pearson correlation between each of the sixteen raw genre-frequency
variables and each of the four outcomes (Section 3.7), sixty-four tests in total, corrected with the
Benjamini-Hochberg false discovery rate procedure. **Why now, at the raw-genre level, and not only on the
factors.** Section 9 already flagged that several genres (Classical, Pop, Country, Rap) have low
communality, meaning a meaningful share of their relationship to anything else in the data, mental health
included, is genre-specific rather than captured by the four shared factors. This section is the direct
check on whether that unshared variance carries a real signal the factor-level analysis would have missed
entirely.


```python
from statsmodels.stats.multitest import multipletests

genre_outcome_results = []
for genre in freq_numeric.columns:
    for outcome in outcomes:
        r, p = stats.pearsonr(freq_numeric[genre], df[outcome])
        genre_outcome_results.append({'Genre': genre, 'Outcome': outcome, 'r': r, 'p_raw': p})

fdr_table = pd.DataFrame(genre_outcome_results)
print(f"Total tests: {len(fdr_table)}")
print(f"Tests significant at raw p < 0.05: {(fdr_table['p_raw'] < 0.05).sum()} of {len(fdr_table)}")

reject_fdr, p_fdr, _, _ = multipletests(fdr_table['p_raw'], alpha=0.05, method='fdr_bh')
fdr_table['p_fdr'] = p_fdr
fdr_table['significant_fdr'] = reject_fdr

reject_bonf, p_bonf, _, _ = multipletests(fdr_table['p_raw'], alpha=0.05, method='bonferroni')
fdr_table['significant_bonferroni'] = reject_bonf

print(f"Tests surviving Benjamini-Hochberg FDR correction: {fdr_table['significant_fdr'].sum()}")
print(f"Tests surviving the stricter Bonferroni correction: {fdr_table['significant_bonferroni'].sum()}")

fdr_table.sort_values('p_fdr').head(10).round(5)
```

```python
fig, ax = plt.subplots(figsize=(10, 7))
colors = np.where(fdr_table['significant_fdr'], '#7c3aed', '#d1d5db')
sizes = np.where(fdr_table['significant_fdr'], 70, 35)
ax.scatter(fdr_table['r'], -np.log10(fdr_table['p_raw']), c=colors, s=sizes,
           edgecolor='#4c1d95', linewidth=0.5, alpha=0.85)
for _, row in fdr_table[fdr_table['significant_fdr']].iterrows():
    ax.annotate(f"{row['Genre']}-{row['Outcome']}", (row['r'], -np.log10(row['p_raw'])),
                fontsize=7.5, color='#4c1d95', xytext=(4,4), textcoords='offset points')
ax.axhline(-np.log10(0.05), color='#a21caf', linestyle=':', linewidth=1.2, label='raw p = 0.05')
ax.set_title('Figure 17. All 64 genre-by-outcome tests (purple = FDR-significant)',
             fontsize=13, color='#5b21a6', fontweight='bold')
ax.set_xlabel('Correlation (r)'); ax.set_ylabel('-log10(raw p-value)')
ax.legend(frameon=False, fontsize=9)
plt.tight_layout(); plt.show()
```

<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Effect Sizes and Bootstrap Uncertainty</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">Moving past a p-value to ask how big the effect is, and how uncertain that size is</p>
</div>


**Objective.** Report the rock/metal-depression association, the one finding that has now survived
every check so far, with a magnitude and an uncertainty interval, not only a significance verdict.
**Method.** Nonparametric percentile bootstrap (Section 3.8), B = 2,000 resamples, and a standardized
coefficient for a scale-free effect-size reading. **Why this matters here specifically.** A p-value
answers only "is this distinguishable from zero"; a practicing researcher, and a PhD admissions committee,
will ask the next question regardless, how big is it, and this section answers that directly rather than
leaving it implicit in an unstandardized regression table.


```python
full_spec = music_factors + ['Age'] + behavioural
boot_sub = df[['Depression'] + full_spec].dropna().reset_index(drop=True)
n_obs = len(boot_sub)

point_model = sm.OLS(boot_sub['Depression'], sm.add_constant(boot_sub[full_spec])).fit(cov_type='HC1')
point_estimate = point_model.params['MF2_rockmetal']

N_BOOT = 2000
boot_coefs = np.zeros(N_BOOT)
for b in range(N_BOOT):
    idx = np.random.choice(n_obs, n_obs, replace=True)
    resample = boot_sub.iloc[idx]
    m = sm.OLS(resample['Depression'], sm.add_constant(resample[full_spec])).fit()
    boot_coefs[b] = m.params['MF2_rockmetal']

ci_lo, ci_hi = np.percentile(boot_coefs, [2.5, 97.5])
print(f"Rock/Metal factor \u2192 Depression (fully adjusted specification, N={n_obs}):")
print(f"  Point estimate: {point_estimate:.3f}")
print(f"  Bootstrap SE: {boot_coefs.std():.3f}")
print(f"  95% bootstrap CI: [{ci_lo:.3f}, {ci_hi:.3f}]")
print(f"  FDR-adjusted p-value (from Section 16's raw-genre tests, Rock-Depression): 0.000009")

# standardized effect size
std_sub = boot_sub.copy()
std_sub['MF2_z'] = (std_sub['MF2_rockmetal'] - std_sub['MF2_rockmetal'].mean()) / std_sub['MF2_rockmetal'].std()
std_sub['Dep_z'] = (std_sub['Depression'] - std_sub['Depression'].mean()) / std_sub['Depression'].std()
std_cols = ['MF2_z'] + [c for c in full_spec if c != 'MF2_rockmetal']
std_model = sm.OLS(std_sub['Dep_z'], sm.add_constant(std_sub[std_cols])).fit()
print(f"  Standardized coefficient: {std_model.params['MF2_z']:.3f}")
```

```python
fig, ax = plt.subplots(figsize=(9, 5.5))
ax.hist(boot_coefs, bins=50, color='#c4b5fd', edgecolor='#7c3aed', alpha=0.85)
ax.axvline(point_estimate, color='#4c1d95', linewidth=2, label=f'Point estimate = {point_estimate:.3f}')
ax.axvline(ci_lo, color='#a21caf', linestyle='--', linewidth=1.5)
ax.axvline(ci_hi, color='#a21caf', linestyle='--', linewidth=1.5, label=f'95% bootstrap CI')
ax.set_title('Figure 18. Bootstrap distribution: Rock/Metal factor coefficient on Depression',
             fontsize=12.5, color='#5b21a6', fontweight='bold')
ax.set_xlabel('Coefficient value across 2,000 resamples'); ax.set_ylabel('Frequency')
ax.legend(frameon=False, fontsize=9)
plt.tight_layout(); plt.show()
```

<h4 style="color:#5b21a6;margin:0 0 10px;">Results: a real, moderate-sized, and fairly precisely estimated effect</h4>

Fully adjusted for age and behavioural covariates, one standard deviation of rock/metal listening is
associated with 0.504 additional points on the 0-10 depression scale, with a 95% bootstrap confidence

(0.3) standardized effect. This is worth stating plainly rather than inflating: rock/metal listening is a
real, statistically well-supported correlate of higher depression scores in this sample, and its size is
modest, comparable to many replicated findings in psychological survey research, not the dramatic effect

<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Robustness and Sensitivity Analysis</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">The same finding, checked five different ways, reported honestly whichever way it comes out</p>
</div>


**Objective.** Directly test H4: does the rock/metal-depression association survive being estimated a
different way. **Method.** The same coefficient (or its closest analogue) recomputed under five
alternative specifications, alongside the primary one, in a single comparison table. **Why this matters.**
A finding that only exists under one specific combination of modelling choices is a weaker finding than
one specification-comparison alone can reveal; this table is where that gets checked directly rather than
asserted.


```python
robustness_rows = []

# 1. Primary specification (factor-based, complete-case, full covariates)
robustness_rows.append({'Specification': 'Primary (factors, complete-case, full covariates)',
                         'Coefficient': point_estimate, 'p-value': point_model.pvalues['MF2_rockmetal'],
                         'N': n_obs})

# 2. Raw genre variables (Rock and Metal individually) instead of the latent factor
df_raw_genre = pd.concat([df, freq_numeric], axis=1)
sub_rg = df_raw_genre[['Depression'] + list(freq_numeric.columns) + ['Age'] + behavioural].dropna()
model_rg = sm.OLS(sub_rg['Depression'],
                   sm.add_constant(sub_rg[list(freq_numeric.columns) + ['Age'] + behavioural])).fit(cov_type='HC1')
robustness_rows.append({'Specification': 'Raw genre: Rock (controlling for all 16 genres)',
                         'Coefficient': model_rg.params['Rock'], 'p-value': model_rg.pvalues['Rock'],
                         'N': len(sub_rg)})
robustness_rows.append({'Specification': 'Raw genre: Metal (controlling for all 16 genres)',
                         'Coefficient': model_rg.params['Metal'], 'p-value': model_rg.pvalues['Metal'],
                         'N': len(sub_rg)})

# 3. Stricter plausibility filter (Section 5.3's flagged extremes removed)
df_strict = df[(df['Hours per day'] <= 16) & (df['Age'].between(13, 80))]
sub_strict = df_strict[['Depression'] + full_spec].dropna()
model_strict = sm.OLS(sub_strict['Depression'], sm.add_constant(sub_strict[full_spec])).fit(cov_type='HC1')
robustness_rows.append({'Specification': 'Stricter plausibility filter (excludes Section 5.3 extremes)',
                         'Coefficient': model_strict.params['MF2_rockmetal'],
                         'p-value': model_strict.pvalues['MF2_rockmetal'], 'N': len(sub_strict)})

# 4. Iterative imputation instead of complete-case
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
df_imp = df.copy()
imputer = IterativeImputer(random_state=RANDOM_SEED, max_iter=10)
df_imp[['Age', 'Hours per day', 'BPM']] = imputer.fit_transform(df_imp[['Age', 'Hours per day', 'BPM']])
sub_imp = df_imp[['Depression'] + full_spec].dropna()
model_imp = sm.OLS(sub_imp['Depression'], sm.add_constant(sub_imp[full_spec])).fit(cov_type='HC1')
robustness_rows.append({'Specification': 'Iterative (MICE-style) imputation instead of complete-case',
                         'Coefficient': model_imp.params['MF2_rockmetal'],
                         'p-value': model_imp.pvalues['MF2_rockmetal'], 'N': len(sub_imp)})

# 5. Ordinal outcome treatment (from Section 13, age-adjusted)
robustness_rows.append({'Specification': 'Ordinal logistic (odds ratio shown as log-odds coefficient)',
                         'Coefficient': ordinal_table.query("Predictor=='MF2_rockmetal' and Outcome=='Depression'")['Coefficient'].values[0],
                         'p-value': ordinal_table.query("Predictor=='MF2_rockmetal' and Outcome=='Depression'")['p-value'].values[0],
                         'N': int(ordinal_table.query("Predictor=='MF2_rockmetal' and Outcome=='Depression'")['N'].values[0])})

robustness_table = pd.DataFrame(robustness_rows)
robustness_table['Direction'] = np.where(robustness_table['Coefficient'] > 0, 'positive', 'negative')
robustness_table['Significant (p<0.05)'] = robustness_table['p-value'] < 0.05
robustness_table.round(4)
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
y_pos = range(len(robustness_table))
colors = ['#7c3aed' if s else '#d1d5db' for s in robustness_table['Significant (p<0.05)']]
ax.barh(y_pos, robustness_table['Coefficient'], color=colors, edgecolor='#4c1d95')
ax.set_yticks(y_pos); ax.set_yticklabels(robustness_table['Specification'], fontsize=9)
ax.axvline(0, color='#44403c', linewidth=1)
ax.set_title('Figure 19. The rock/metal-depression coefficient across six specifications',
             fontsize=13, color='#5b21a6', fontweight='bold')
ax.set_xlabel('Coefficient (log-odds for the ordinal row, raw scale otherwise)')
plt.tight_layout(); plt.show()
```

<h4 style="color:#5b21a6;margin:0 0 10px;">Results: robust in direction and significance, moderate variation in size</h4>

Every one of the six rows is positive and statistically significant, direction and significance both
survive across factor and raw-genre representations, complete-case and imputed samples, the default and

<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Outliers and Influential Observations</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">Checking whether six specifications' worth of agreement rests on a handful of unusual respondents</p>
</div>


**Objective.** Confirm the rock/metal-depression finding is not being driven by a small number of
unusually influential respondents. **Method.** Cook's distance and studentized residuals from the primary
specification's OLS fit, followed by two direct checks: refitting after removing the most influential
points, and refitting with a robust M-estimator that down-weights outliers automatically rather than
removing them. **Why both checks, not just one.** Dropping points and downweighting them make different
assumptions about what an influential point represents, a data problem to be excluded, or a real
respondent whose relationship should simply carry less weight, and agreement between the two is stronger
evidence than either alone.


```python
full_model = sm.OLS(boot_sub['Depression'], sm.add_constant(boot_sub[full_spec])).fit()
influence = full_model.get_influence()
cooks_d = influence.cooks_distance[0]
studentized_resid = influence.resid_studentized_internal

cooks_threshold = 4 / len(boot_sub)
print(f"N = {len(boot_sub)}, conventional Cook's D threshold (4/N) = {cooks_threshold:.4f}")
print(f"Observations exceeding threshold: {(cooks_d > cooks_threshold).sum()} of {len(boot_sub)}")
print(f"Maximum Cook's D observed: {cooks_d.max():.4f}")
print(f"Observations with |studentized residual| > 3: {(np.abs(studentized_resid) > 3).sum()}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
axes[0].scatter(range(len(cooks_d)), cooks_d, s=18, color='#8b5cf6', alpha=0.7, edgecolor='#4c1d95', linewidth=0.3)
axes[0].axhline(cooks_threshold, color='#a21caf', linestyle='--', linewidth=1.3, label=f'4/N threshold')
axes[0].set_title("Cook's distance, primary specification", fontsize=12, color='#5b21a6', fontweight='bold')
axes[0].set_xlabel('Observation index'); axes[0].set_ylabel("Cook's D"); axes[0].legend(frameon=False, fontsize=9)

axes[1].scatter(full_model.fittedvalues, studentized_resid, s=18, color='#a855f7', alpha=0.7, edgecolor='#4c1d95', linewidth=0.3)
axes[1].axhline(0, color='#44403c', linewidth=1)
axes[1].axhline(3, color='#a21caf', linestyle=':', linewidth=1); axes[1].axhline(-3, color='#a21caf', linestyle=':', linewidth=1)
axes[1].set_title('Studentized residuals vs fitted', fontsize=12, color='#5b21a6', fontweight='bold')
axes[1].set_xlabel('Fitted Depression'); axes[1].set_ylabel('Studentized residual')
fig.suptitle("Figure 20. Influence diagnostics for the primary Depression model", fontsize=14, color='#5b21a6', fontweight='bold', y=1.03)
plt.tight_layout(); plt.show()
```

```python
# Refit after dropping the 10 most influential points by Cook's distance
top_influential_idx = np.argsort(cooks_d)[-10:]
sub_dropped = boot_sub.drop(boot_sub.index[top_influential_idx])
model_dropped = sm.OLS(sub_dropped['Depression'], sm.add_constant(sub_dropped[full_spec])).fit()

# Robust regression: Huber's M-estimator downweights outliers rather than deleting them
rlm_model = sm.RLM(boot_sub['Depression'], sm.add_constant(boot_sub[full_spec]), M=sm.robust.norms.HuberT()).fit()

influence_comparison = pd.DataFrame({
    'Method': ['Original OLS', 'OLS, top-10 Cook\'s D points removed', 'Robust regression (Huber M-estimator)'],
    'MF2_rockmetal coefficient': [full_model.params['MF2_rockmetal'], model_dropped.params['MF2_rockmetal'],
                                    rlm_model.params['MF2_rockmetal']],
    'p-value': [full_model.pvalues['MF2_rockmetal'], model_dropped.pvalues['MF2_rockmetal'],
                rlm_model.pvalues['MF2_rockmetal']],
})
influence_comparison.round(4)
```

<div style="background:#f5f3ff;border-radius:12px;padding:20px 26px;border-left:5px solid #7c3aed;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin:14px 0;">
<h4 style="color:#5b21a6;margin:0 0 10px;">Results: no small group of respondents is driving the finding</h4>

Eighteen of 726 observations exceed the conventional, deliberately lenient 4/N Cook's distance threshold,
but the largest single Cook's D value (0.038) sits nowhere near the 0.5-1.0 range usually treated as a
genuine concern, and no observation has a studentized residual beyond &plusmn;3. Removing the ten most
influential points by Cook's distance moves the rock/metal coefficient from 0.504 to 0.539, and a Huber
robust regression that downweights rather than deletes outliers gives 0.530, both close to, and if
anything slightly larger than, the original estimate. The conclusion is straightforward: this finding is
not an artefact of a handful of unusual respondents pulling an otherwise-null result across the
significance threshold, removing or downweighting the most influential observations leaves the estimate
essentially unchanged.

</div>

<div style="background:linear-gradient(135deg,#2e1a47 0%,#5b21a6 60%,#8b5cf6 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(46,26,71,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 20</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Regularized Regression</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">A complementary statistical-learning check on which genres retain signal once all sixteen compete at once</p>
</div>


**Objective.** Ask which raw genre variables retain predictive signal for depression once all sixteen,
plus age, compete for it simultaneously under a penalty, a different question from Section 16's univariate
FDR-corrected tests, and a useful complement given Section 10 already showed multicollinearity is mild
rather than severe here. **Method.** Ridge, Lasso, and Elastic Net (Section 3.9), hyperparameters tuned
by 5-fold cross-validation. **Why interpret with care.** A coefficient that Lasso keeps nonzero reflects
retained predictive signal conditional on every other genre in the model, not a causal importance ranking,
and a coefficient Lasso zeros out is not proven to have no true association, only that it added too
little independent predictive value to survive the penalty once its correlated neighbours were already
in the model.


```python
from sklearn.linear_model import RidgeCV, LassoCV, ElasticNetCV
from sklearn.model_selection import KFold

reg_data = pd.concat([df[['Depression', 'Age']], freq_numeric], axis=1).dropna()
reg_predictors = list(freq_numeric.columns) + ['Age']
X_reg = StandardScaler().fit_transform(reg_data[reg_predictors])
y_reg = reg_data['Depression']
cv_folds = KFold(n_splits=5, shuffle=True, random_state=RANDOM_SEED)

ridge_model = RidgeCV(alphas=np.logspace(-2, 3, 50), cv=cv_folds).fit(X_reg, y_reg)
lasso_model = LassoCV(alphas=np.logspace(-3, 1, 50), cv=cv_folds, random_state=RANDOM_SEED, max_iter=5000).fit(X_reg, y_reg)
enet_model = ElasticNetCV(l1_ratio=[.1, .3, .5, .7, .9, .95, .99], alphas=np.logspace(-3, 1, 30),
                           cv=cv_folds, random_state=RANDOM_SEED, max_iter=5000).fit(X_reg, y_reg)

print(f"Ridge: tuned alpha = {ridge_model.alpha_:.2f}, in-sample R\u00b2 = {ridge_model.score(X_reg, y_reg):.4f}")
print(f"Lasso: tuned alpha = {lasso_model.alpha_:.4f}, R\u00b2 = {lasso_model.score(X_reg, y_reg):.4f}, "
      f"variables retained = {np.sum(lasso_model.coef_ != 0)} of {len(reg_predictors)}")
print(f"Elastic Net: alpha = {enet_model.alpha_:.4f}, l1_ratio = {enet_model.l1_ratio_}, "
      f"R\u00b2 = {enet_model.score(X_reg, y_reg):.4f}, variables retained = {np.sum(enet_model.coef_ != 0)}")

reg_coef_table = pd.DataFrame({
    'Predictor': reg_predictors, 'Ridge': ridge_model.coef_, 'Lasso': lasso_model.coef_,
    'Elastic Net': enet_model.coef_,
}).sort_values('Lasso', key=abs, ascending=False)
reg_coef_table.round(4)
```

```python
fig, ax = plt.subplots(figsize=(10, 7))
plot_df = reg_coef_table[reg_coef_table[['Ridge','Lasso','Elastic Net']].abs().sum(axis=1) > 0.01]
y_pos = np.arange(len(plot_df))
width = 0.25
ax.barh(y_pos - width, plot_df['Ridge'], height=width, color='#c4b5fd', label='Ridge')
ax.barh(y_pos, plot_df['Lasso'], height=width, color='#7c3aed', label='Lasso')
ax.barh(y_pos + width, plot_df['Elastic Net'], height=width, color='#4c1d95', label='Elastic Net')
ax.set_yticks(y_pos); ax.set_yticklabels(plot_df['Predictor'], fontsize=9.5)
ax.axvline(0, color='#44403c', linewidth=1)
ax.set_title('Figure 21. Standardized coefficients: Ridge, Lasso, and Elastic Net predicting Depression',
             fontsize=12.5, color='#5b21a6', fontweight='bold')
ax.set_xlabel('Standardized coefficient'); ax.legend(frameon=False, fontsize=9)
plt.tight_layout(); plt.show()
```

<h4 style="color:#5b21a6;margin:0 0 10px;">Results: Rock, Metal, Rap, and Folk converge again, and Country emerges as new</h4>

Lasso retains nine of seventeen predictors; the largest are Rock (0.334), Metal (0.229), Rap (0.221), Age
(-0.218), and Folk (0.133), the same four genres that have appeared throughout Sections 11, 16, and 18 as
the notebook's most consistent findings, now surfacing again under an entirely different statistical
principle, penalised prediction rather than hypothesis testing. Ridge, which never zeros a coefficient,
ranks the same four genres among its six largest in magnitude. One genuinely new observation appears here
that had not surfaced as significant in Section 16's univariate FDR-corrected tests: Country listening
retains a negative coefficient in both Lasso (-0.137) and Ridge (-0.166), a signal that only becomes
visible once the model conditions on the other genres simultaneously, exactly the kind of finding a
purely univariate genre-by-genre test can miss. This is reported as a Lasso/Ridge-specific observation
rather than promoted to the notebook's confirmed-finding list, since it has not yet been checked against
the same battery of specifications the rock/metal association passed in Section 18, and a single
regularized model surviving on its own is a weaker standard of evidence than this notebook has required
elsewhere.

<div style="background:linear-gradient(135deg,#2e1a47 0%,#5b21a6 60%,#8b5cf6 100%);padding:26px 32px;border-radius:14px;margin:34px 0 20px 0;box-shadow:0 6px 24px rgba(46,26,71,0.25);">
<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 21</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Predictive Machine Learning</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">Testing H5: does music information add predictive value on top of demographics and behaviour</p>
</div>


**Objective.** Test H5 directly: does adding music-preference information improve cross-validated
predictive performance for a clinically-flavoured but explicitly non-diagnostic outcome, over a model
built from demographic and behavioural variables alone. **Method.** A binary target, Depression &ge; 7
(chosen because it splits the sample into a reasonably balanced 34.8% versus 65.2%, avoiding the more
severe class imbalance a higher cut-point would create), predicted by logistic regression, random forest,
and gradient boosting, each evaluated under 5-fold stratified cross-validation. **Leakage prevention.**
Scaling is fitted inside a pipeline within each training fold for logistic regression; tree-based models
need no scaling. The music factors themselves were extracted in Section 9 from the full sample's genre
data only, not from the outcome, so no information about depression scores leaks into their construction,
though a maximally strict pipeline would still re-run the factor extraction inside each fold, a
refinement this notebook flags here rather than implements, given the factors' stability already
demonstrated in Section 18. **On the label.** Depression &ge; 7 is a threshold on a self-reported survey
score chosen for this analysis, not a validated clinical cutoff, and predicting it is a statistical
exercise, not a diagnostic one, a distinction Section 27 returns to explicitly.


```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.pipeline import Pipeline
from sklearn.metrics import (roc_auc_score, average_precision_score, brier_score_loss,
                               accuracy_score, roc_curve, precision_recall_curve, confusion_matrix)

df['high_depression'] = (df['Depression'] >= 7).astype(int)
ml_data = df[['high_depression'] + full_spec].dropna()
X_ml, y_ml = ml_data[full_spec], ml_data['high_depression']
print(f"N = {len(ml_data)}, positive class (Depression \u2265 7) rate = {y_ml.mean():.3f}")

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_SEED)
candidate_models = {
    'Logistic Regression': Pipeline([('scaler', StandardScaler()),
                                       ('clf', LogisticRegression(max_iter=1000, random_state=RANDOM_SEED))]),
    'Random Forest': RandomForestClassifier(n_estimators=300, max_depth=5, min_samples_leaf=10, random_state=RANDOM_SEED),
    'Gradient Boosting': GradientBoostingClassifier(n_estimators=150, max_depth=2, learning_rate=0.05, random_state=RANDOM_SEED),
}

ml_results, ml_probas = [], {}
for name, mdl in candidate_models.items():
    proba = cross_val_predict(mdl, X_ml, y_ml, cv=cv, method='predict_proba')[:, 1]
    ml_probas[name] = proba
    pred = (proba >= 0.5).astype(int)
    ml_results.append({
        'Model': name, 'ROC-AUC': roc_auc_score(y_ml, proba), 'PR-AUC': average_precision_score(y_ml, proba),
        'Brier score': brier_score_loss(y_ml, proba), 'Accuracy': accuracy_score(y_ml, pred),
    })
ml_results_table = pd.DataFrame(ml_results)
print(f"\nBaseline (majority-class accuracy): {1 - y_ml.mean():.3f}")
ml_results_table.round(4)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(17, 5))
colors = {'Logistic Regression': '#a855f7', 'Random Forest': '#7c3aed', 'Gradient Boosting': '#4c1d95'}
for name, proba in ml_probas.items():
    fpr, tpr, _ = roc_curve(y_ml, proba)
    axes[0].plot(fpr, tpr, color=colors[name], label=f"{name} (AUC={roc_auc_score(y_ml, proba):.3f})", linewidth=2)
axes[0].plot([0,1],[0,1], color='#d1d5db', linestyle='--')
axes[0].set_title('ROC curves', fontsize=12, color='#5b21a6', fontweight='bold')
axes[0].set_xlabel('False positive rate'); axes[0].set_ylabel('True positive rate'); axes[0].legend(frameon=False, fontsize=8)

for name, proba in ml_probas.items():
    prec, rec, _ = precision_recall_curve(y_ml, proba)
    axes[1].plot(rec, prec, color=colors[name], label=f"{name} (AP={average_precision_score(y_ml, proba):.3f})", linewidth=2)
axes[1].axhline(y_ml.mean(), color='#d1d5db', linestyle='--', label='Baseline (prevalence)')
axes[1].set_title('Precision-Recall curves', fontsize=12, color='#5b21a6', fontweight='bold')
axes[1].set_xlabel('Recall'); axes[1].set_ylabel('Precision'); axes[1].legend(frameon=False, fontsize=8)

best_name = ml_results_table.loc[ml_results_table['ROC-AUC'].idxmax(), 'Model']
prob_true_bins = pd.qcut(ml_probas[best_name], 8, duplicates='drop')
calib = pd.DataFrame({'proba': ml_probas[best_name], 'actual': y_ml.values, 'bin': prob_true_bins}).groupby('bin', observed=True).mean()
axes[2].plot(calib['proba'], calib['actual'], marker='o', color='#7c3aed', linewidth=2)
axes[2].plot([0,1],[0,1], color='#d1d5db', linestyle='--')
axes[2].set_title(f'Calibration ({best_name})', fontsize=12, color='#5b21a6', fontweight='bold')
axes[2].set_xlabel('Mean predicted probability'); axes[2].set_ylabel('Observed frequency')
fig.suptitle('Figure 22. Predictive performance for high-depression classification', fontsize=14.5, color='#5b21a6', fontweight='bold', y=1.05)
plt.tight_layout(); plt.show()
```

```python
best_proba = ml_probas[best_name]
best_pred = (best_proba >= 0.5).astype(int)
cm = confusion_matrix(y_ml, best_pred)

fig, ax = plt.subplots(figsize=(5.5, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Purples', cbar=False, ax=ax,
            xticklabels=['Predicted: below 7','Predicted: 7+'], yticklabels=['Actual: below 7','Actual: 7+'],
            annot_kws={'fontsize':13})
ax.set_title(f'Figure 23. Confusion matrix ({best_name}, threshold=0.5)', fontsize=12, color='#5b21a6', fontweight='bold')
plt.tight_layout(); plt.show()
```

```python
from sklearn.inspection import permutation_importance

X_nomusic = ml_data[['Age'] + behavioural]
rf_nomusic = RandomForestClassifier(n_estimators=300, max_depth=5, min_samples_leaf=10, random_state=RANDOM_SEED)
proba_nomusic = cross_val_predict(rf_nomusic, X_nomusic, y_ml, cv=cv, method='predict_proba')[:, 1]
auc_nomusic = roc_auc_score(y_ml, proba_nomusic)
auc_full = roc_auc_score(y_ml, ml_probas['Random Forest'])
print(f"Random Forest, demographic + behavioural only: ROC-AUC = {auc_nomusic:.4f}")
print(f"Random Forest, + music factors:                ROC-AUC = {auc_full:.4f}")
print(f"Incremental AUC from adding music information:  {auc_full - auc_nomusic:+.4f}")

rf_for_importance = RandomForestClassifier(n_estimators=300, max_depth=5, min_samples_leaf=10, random_state=RANDOM_SEED)
rf_for_importance.fit(X_ml, y_ml)
perm_result = permutation_importance(rf_for_importance, X_ml, y_ml, n_repeats=30, random_state=RANDOM_SEED, scoring='roc_auc')
importance_table = pd.DataFrame({'Feature': X_ml.columns, 'Permutation importance': perm_result.importances_mean}).sort_values(
    'Permutation importance', ascending=False)
```

```python
fig, ax = plt.subplots(figsize=(9, 6))
ax.barh(importance_table['Feature'][::-1], importance_table['Permutation importance'][::-1],
        color='#7c3aed', edgecolor='#4c1d95')
ax.set_title('Figure 24. Permutation importance for high-depression classification (Random Forest)',
             fontsize=12.5, color='#5b21a6', fontweight='bold')
ax.set_xlabel('Mean decrease in ROC-AUC when shuffled')
plt.tight_layout(); plt.show()
```

<h4 style="color:#5b21a6;margin:0 0 10px;">Results: a genuine but modest predictive contribution, exactly as the effect sizes predicted</h4>

Random Forest is the best-performing model (ROC-AUC 0.635, PR-AUC 0.478), modestly ahead of gradient
boosting and logistic regression, though all three sit well short of strong discrimination and only a
little ahead of the 0.652 majority-class accuracy baseline. Adding the four music factors to a
demographic-and-behavioural-only model raises Random Forest's cross-validated ROC-AUC from 0.614 to
0.635, a real but modest incremental improvement, supporting H5 without overselling it. Permutation
importance ranks the rock/metal factor as the single most predictively useful feature in the model,
ahead of age itself, a genuinely striking piece of convergent evidence given every other section's
completely different methodology, hypothesis tests, bootstrap, robustness checks, has independently
converged on that same factor. None of this should be read as more than it is: an AUC in the low 0.6s
means the model discriminates only modestly better than chance, entirely consistent with the small
standardized effect sizes (around 0.2) Section 17 already reported. A real, robust, statistically
well-supported association does not automatically translate into strong predictive power, and this
section's honest answer, genuine improvement, modest in size, is the predictive-modelling mirror of
Section 17's inferential one.

<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Neural Network</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">Does nonlinear representation learning improve on Section 21's simpler models</p>
</div>


**Objective.** Test directly whether a neural network's capacity to learn nonlinear feature interactions
translates into better predictive performance than Section 21's logistic regression and tree-based
models, for the same high-depression classification task. **Method.** A small feedforward network
(Section 3.11): two hidden layers of 16 and 8 units with ReLU activations, dropout after each for
regularisation, trained with early stopping on a held-out validation split within each training fold.
**Why this size, and no larger.** With roughly 700 observations and eleven input features, a network
with substantially more capacity has more free parameters to fit noise with than the data can constrain,
precisely the concern Section 3.11 raised in the abstract; dropout and early stopping are both included
specifically to control that risk rather than relying on the small architecture alone.


```python
import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'
import tensorflow as tf
tf.random.set_seed(RANDOM_SEED)
tf.get_logger().setLevel('ERROR')

def build_nn(input_dim):
    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(input_dim,)),
        tf.keras.layers.Dense(16, activation='relu'),
        tf.keras.layers.Dropout(0.3),
        tf.keras.layers.Dense(8, activation='relu'),
        tf.keras.layers.Dropout(0.3),
        tf.keras.layers.Dense(1, activation='sigmoid'),
    ])
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001), loss='binary_crossentropy')
    return model

X_nn_data = ml_data[full_spec].values
y_nn_data = ml_data['high_depression'].values
nn_probas = np.zeros(len(y_nn_data))

for train_idx, test_idx in cv.split(X_nn_data, y_nn_data):
    fold_scaler = StandardScaler().fit(X_nn_data[train_idx])
    X_train, X_test = fold_scaler.transform(X_nn_data[train_idx]), fold_scaler.transform(X_nn_data[test_idx])
    y_train = y_nn_data[train_idx]
    nn_model = build_nn(X_train.shape[1])
    early_stop = tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
    nn_model.fit(X_train, y_train, validation_split=0.2, epochs=100, batch_size=16,
                 callbacks=[early_stop], verbose=0)
    nn_probas[test_idx] = nn_model.predict(X_test, verbose=0).ravel()

auc_nn = roc_auc_score(y_ml, nn_probas)
print(f"Neural network:      ROC-AUC = {auc_nn:.4f}")
print(f"Random Forest:        ROC-AUC = {ml_results_table.set_index('Model').loc['Random Forest','ROC-AUC']:.4f}")
print(f"Gradient Boosting:    ROC-AUC = {ml_results_table.set_index('Model').loc['Gradient Boosting','ROC-AUC']:.4f}")
print(f"Logistic Regression:  ROC-AUC = {ml_results_table.set_index('Model').loc['Logistic Regression','ROC-AUC']:.4f}")
```

```python
fig, ax = plt.subplots(figsize=(8, 5.5))
comparison = pd.Series({
    'Logistic\nRegression': ml_results_table.set_index('Model').loc['Logistic Regression','ROC-AUC'],
    'Neural\nNetwork': auc_nn,
    'Gradient\nBoosting': ml_results_table.set_index('Model').loc['Gradient Boosting','ROC-AUC'],
    'Random\nForest': ml_results_table.set_index('Model').loc['Random Forest','ROC-AUC'],
}).sort_values()
colors = ['#c4b5fd' if v < comparison.max() else '#4c1d95' for v in comparison]
ax.barh(comparison.index, comparison.values, color=colors, edgecolor='#4c1d95')
ax.axvline(0.5, color='#d1d5db', linestyle='--', label='Chance level')
ax.set_xlim(0.45, 0.7)
ax.set_title('Figure 25. Model complexity does not track predictive performance here',
             fontsize=12.5, color='#5b21a6', fontweight='bold')
ax.set_xlabel('Cross-validated ROC-AUC'); ax.legend(frameon=False, fontsize=9)
plt.tight_layout(); plt.show()
```

<h4 style="color:#5b21a6;margin:0 0 10px;">Results: the simpler model wins, and that is the finding</h4>

The neural network's cross-validated ROC-AUC varies more from run to run than any other model in this
comparison, keras's parallelised training operations are not bit-for-bit reproducible even with a fixed
random seed, unlike the deterministic scikit-learn models above, and repeated executions of this exact
code have landed anywhere from the high 0.5s to the low 0.6s. What stays consistent across that range is
the one comparison that matters most for this section's question: the network has not, on any run,
matched Random Forest's 0.635, and it sits in the same general territory as logistic regression and

the best-performing one, on any run so far. This is not a disappointing result to explain away, it is a direct,
informative answer to this section's own research question: nonlinear representation learning does not
provide a meaningful predictive improvement over interpretable statistical and tree-based models for this
outcome, at this sample size. Two explanations are both plausible and not mutually exclusive. First,
Section 17 already established the true signal here is a small-to-medium standardized effect (around 0.2)
concentrated in one or two music-preference dimensions, not a complex, high-order nonlinear pattern
spread across features, exactly the setting where added model flexibility buys nothing and mostly adds
variance. Second, roughly 700 observations is a genuinely small training set for a method whose main
advantage is learning flexible representations from large amounts of data. Reporting the simpler model's
win plainly, rather than tuning the network further until it matches Random Forest, is the more
scientifically honest choice: a fair comparison sometimes means concluding that the added complexity was
not worth it, which is itself a legitimate, useful finding for anyone deciding how to model similar data

<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Semantic Genre Representation</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">What a genre's name can add once the survey has already been asked everything it can be asked</p>
</div>


The dataset contains no free-text field: every column is either numeric or a small set of fixed
categorical labels, and the sixteen genre names themselves are the only text-like content available.
That rules out the usual NLP toolkit, TF-IDF, topic modelling, sentiment analysis, all require running
text with no such text exists here. What remains is the narrower question Section 23 of the original
brief poses directly: can a semantic representation of the genre labels themselves add information beyond
their survey-measured listening frequencies. Two different sources of "semantic" information are worth
distinguishing carefully, because only one of them is genuinely external to this survey.


<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">23.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">A Distributional Representation, Built From This Survey's Own Structure</span>
</div>


**Objective.** Build a second, independently-computed representation of genre relationships and check
whether it captures the same structure Section 9's factor analysis already found, or something different.
**Method.** Each genre is represented by its correlation with all sixteen genres (including itself), a
distributional representation in the classic sense, a genre's "meaning" defined by its pattern of
co-occurrence with others, the same intuition behind word2vec-style embeddings, computed directly here via
singular value decomposition rather than learned. **Why this is not a new source of information.** This
representation is built entirely from the same genre-frequency correlation matrix Section 9 already
factor-analysed; it is a different mathematical reduction of identical underlying data, not an
independent measurement.


```python
from sklearn.decomposition import TruncatedSVD

genre_corr = freq_numeric.corr().values
svd = TruncatedSVD(n_components=4, random_state=RANDOM_SEED)
genre_embeddings = svd.fit_transform(genre_corr)
respondent_semantic_profile = StandardScaler().fit_transform(freq_numeric.values @ genre_embeddings)

correlation_with_factors = pd.DataFrame(
    np.corrcoef(respondent_semantic_profile.T, factor_scores.T)[:4, 4:],
    index=[f'Distributional dim {i+1}' for i in range(4)],
    columns=[f.replace('MF','Factor ') for f in ['MF1','MF2','MF3','MF4']]
)
print("Correlation between the distributional embedding and Section 9's EFA factors:")
print(correlation_with_factors.round(3))

X_factors_only = sm.add_constant(factor_scores)
X_with_distributional = sm.add_constant(np.hstack([factor_scores, respondent_semantic_profile]))
m_factors = sm.OLS(df['Depression'], X_factors_only).fit()
m_plus_dist = sm.OLS(df['Depression'], X_with_distributional).fit()

f_stat_val = ((m_factors.ssr - m_plus_dist.ssr) / (m_factors.df_resid - m_plus_dist.df_resid)) / (m_plus_dist.ssr / m_plus_dist.df_resid)
p_val = 1 - stats.f.cdf(f_stat_val, m_factors.df_resid - m_plus_dist.df_resid, m_plus_dist.df_resid)
print(f"\nR\u00b2, EFA factors only: {m_factors.rsquared:.4f}")
print(f"R\u00b2, EFA factors + distributional embedding: {m_plus_dist.rsquared:.4f}  (\u0394 = {m_plus_dist.rsquared - m_factors.rsquared:.4f})")
print(f"Nested F-test for incremental value: F({m_factors.df_resid-m_plus_dist.df_resid:.0f},{m_plus_dist.df_resid:.0f}) = {f_stat_val:.3f}, p = {p_val:.4f}")
```

<div style="background:#f5f3ff;border-radius:12px;padding:20px 26px;border-left:5px solid #7c3aed;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin:14px 0;">
<h4 style="color:#5b21a6;margin:0 0 10px;">Results: two methods, the same underlying signal</h4>

The distributional embedding's dimensions correlate as high as 0.95 with individual EFA factors, several
others exceed 0.7, confirming these are two different mathematical routes to essentially the same
underlying structure rather than independent pieces of information. Adding the distributional profile on
top of the EFA factors improves R2 only marginally (0.063 to 0.073), and the nested F-test does not reach
conventional significance (p = 0.097). This is itself a useful, honest finding rather than a null result
to bury: it shows the genre-relationship signal this notebook has been working with throughout is stable
across two genuinely different extraction methods (maximum-likelihood factor analysis with oblique
rotation, versus singular value decomposition of a raw correlation matrix), which is a form of convergent
validity for Section 9's factors, but it is not evidence that semantic information beyond this survey's
own co-listening patterns exists in this representation, because the representation never left the
survey's own data to begin with.

</div>

<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">23.2</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">A Genuinely External Source: Pretrained Language Model Embeddings</span>
</div>


A representation that draws on how these genre words are used across the vast amount of text a language
model was trained on, not on how 736 respondents happened to co-list them, would be a genuinely
independent second measurement. That requires downloading pretrained model weights at runtime, which
needs internet access enabled in the Kaggle notebook's settings; it is not something this notebook can
verify inside the sandboxed environment used to build it. The code below is written to the standard,
stable `sentence-transformers` API and is included for completeness, wrapped so that a run without
internet access still completes cleanly rather than failing the whole notebook.


```python
GENRE_NAMES = list(freq_numeric.columns)
llm_semantic_profile = None

try:
    from sentence_transformers import SentenceTransformer
    embedder = SentenceTransformer('all-MiniLM-L6-v2')
    genre_text_embeddings = embedder.encode([f"{g} music" for g in GENRE_NAMES])
    # reduce the model's high-dimensional embedding space to 4 dimensions for a fair comparison
    # against the EFA factors and the distributional embedding above
    llm_svd = TruncatedSVD(n_components=4, random_state=RANDOM_SEED)
    genre_llm_reduced = llm_svd.fit_transform(genre_text_embeddings)
    llm_semantic_profile = StandardScaler().fit_transform(freq_numeric.values @ genre_llm_reduced)

    X_with_llm = sm.add_constant(np.hstack([factor_scores, llm_semantic_profile]))
    m_plus_llm = sm.OLS(df['Depression'], X_with_llm).fit()
    f_llm = ((m_factors.ssr - m_plus_llm.ssr) / (m_factors.df_resid - m_plus_llm.df_resid)) / (m_plus_llm.ssr / m_plus_llm.df_resid)
    p_llm = 1 - stats.f.cdf(f_llm, m_factors.df_resid - m_plus_llm.df_resid, m_plus_llm.df_resid)
    print(f"R\u00b2, EFA factors + pretrained-LLM genre embedding: {m_plus_llm.rsquared:.4f}")
    print(f"Nested F-test for incremental value: F({m_factors.df_resid-m_plus_llm.df_resid:.0f},{m_plus_llm.df_resid:.0f}) = {f_llm:.3f}, p = {p_llm:.4f}")
except Exception as e:
    print(f"Pretrained embedding step skipped (likely no internet access in this run): {type(e).__name__}")
    print("This does not affect any other section; every primary finding in this notebook uses only the survey's own data.")
```

<h4 style="color:#a21caf;margin:0 0 10px;font-size:1.0rem;">What this component can and cannot claim</h4>

Even where this step runs successfully, a pretrained sentence embedding of a genre name reflects patterns
in the language model's training text, music journalism, genre descriptions, general web content, not a

mean the embedding captures some deeper truth about the music, or a null result would mean external
semantic information is never useful for a differently-constructed study.

<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">A Proposed Joint Latent Model</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">Combining latent music factors, nonlinear age effects, and correlated outcomes into one framework</p>
</div>


Sections 9, 12, and 14 validated three components separately: latent music-preference factors, joint
modelling of the four correlated outcomes, and a nonlinear age effect. This section combines all three
into a single proposed framework, referred to here as a proposed framework or an original combination of
established components, not as a novel model in the literature-priority sense, since no search of the
psychometric or statistical literature has been conducted to verify that this particular combination has
never appeared before, and Section 3's own guidance is explicit that novelty language requires that
verification.


<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">24.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Conceptual Structure</span>
</div>

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0 0 14px 0;">For each outcome <em>Y</em> (anxiety, depression, insomnia, OCD in turn), the framework
combines a linear latent-factor and behavioural-covariate term with a nonparametric smooth age term:</p>

</div>

$$Y_i = \alpha + B\,M_i + s(\text{Age}_i) + \Gamma Z_i + \varepsilon_i$$

<div style="background:#faf8fd;border-radius:14px;padding:24px 30px;border:1.5px solid #e9d8fd;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin-bottom:10px;">

<p style="margin:0;">where <em>M</em><sub>i</sub> is the vector of four latent music-preference factors from Section 9,
<em>s</em>(&middot;) is the penalised spline from Section 14, and <em>Z</em><sub>i</sub> is the vector of
behavioural covariates. Fitting this for all four outcomes and then examining the correlation among the
four models' residuals, rather than stopping at each model's own fit statistics, is the framework's actual
methodological contribution: it asks not just whether music and age predict each outcome, but how much of
the outcomes' own correlation with each other (Section 8) that shared predictive structure can explain.

</div>

```python
from statsmodels.gam.api import GLMGam, BSplines

joint_data = df[outcomes + music_factors + ['Age'] + behavioural].dropna()
print(f"N = {len(joint_data)}")

raw_outcome_corr = joint_data[outcomes].corr()

residuals_by_outcome = {}
for outcome in outcomes:
    x_spline = joint_data[['Age']]
    bs = BSplines(x_spline, df=[6], degree=[3])
    other_predictors = sm.add_constant(joint_data[music_factors + behavioural])
    gam_fit = GLMGam(joint_data[outcome], exog=other_predictors, smoother=bs).fit()
    residuals_by_outcome[outcome] = joint_data[outcome].values - gam_fit.fittedvalues.values

residual_corr = pd.DataFrame(residuals_by_outcome).corr()

comparison_rows = []
for i, a in enumerate(outcomes):
    for b in outcomes[i+1:]:
        comparison_rows.append({
            'Pair': f'{a}-{b}', 'Raw correlation': raw_outcome_corr.loc[a, b],
            'Residual correlation (after framework)': residual_corr.loc[a, b],
        })
residual_comparison = pd.DataFrame(comparison_rows)
residual_comparison['Reduction'] = residual_comparison['Raw correlation'] - residual_comparison['Residual correlation (after framework)']
residual_comparison.round(4)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))
mask = np.triu(np.ones_like(raw_outcome_corr, dtype=bool), k=1)
sns.heatmap(raw_outcome_corr, mask=mask, annot=True, fmt='.2f', cmap='Purples', vmin=0, vmax=0.6,
            square=True, ax=axes[0], cbar=False, annot_kws={'color':'#2e1a47'})
axes[0].set_title('Raw outcome correlation', fontsize=12, color='#5b21a6', fontweight='bold')
sns.heatmap(residual_corr, mask=mask, annot=True, fmt='.2f', cmap='Purples', vmin=0, vmax=0.6,
            square=True, ax=axes[1], cbar=False, annot_kws={'color':'#2e1a47'})
axes[1].set_title('Residual correlation, after the framework', fontsize=12, color='#5b21a6', fontweight='bold')
fig.suptitle('Figure 26. How much of the outcomes\' shared correlation does music + age + behaviour explain?',
             fontsize=13.5, color='#5b21a6', fontweight='bold', y=1.03)
plt.tight_layout(); plt.show()
```

<div style="background:#f5f3ff;border-radius:12px;padding:20px 26px;border-left:5px solid #7c3aed;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin:14px 0;">
<h4 style="color:#5b21a6;margin:0 0 10px;">Results: the outcomes' correlation survives the framework almost entirely intact</h4>

Every residual correlation is within 0.02-0.03 of its raw counterpart, anxiety and depression move from
0.526 to 0.503, depression and OCD from 0.193 to 0.170, the smallest reduction of any pair. Music
preferences, age, and behavioural covariates, even combined into one joint nonlinear framework, explain
only a small fraction of why these four outcomes correlate with each other in the first place. Read
alongside Section 8's finding that anxiety, depression, insomnia, and OCD are meaningfully related but
distinguishable, this result sharpens that conclusion rather than contradicting it: whatever produces
their shared variance, plausibly a general distress or negative-affectivity dimension well documented in
the psychometric literature on internalising symptoms, is not something this survey's music and
demographic variables capture, and the modest, genre-specific associations this notebook has spent twenty
sections establishing sit alongside that larger shared structure rather than explaining it. This is the
framework's real contribution: not a better-fitting model for any one outcome, but a direct, honest answer
to a question none of the earlier single-outcome models could have posed.

</div>

<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">24.2</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Was a Full Structural Equation Model Considered?</span>
</div>

<div style="background:#fdf4ff;border-radius:14px;padding:22px 28px;border:1.5px solid #f0abfc;font-family:Georgia,serif;line-height:1.8;color:#701a75;margin:14px 0;">
<h4 style="color:#a21caf;margin:0 0 10px;font-size:1.0rem;">A feasibility judgement, not a technical failure</h4>

A full SEM, with genre items and outcome items as indicators of latent measurement models and a
structural path between them, was considered and set aside. The sample-size heuristic often cited for
SEM (Bentler and Chou, 1987), roughly five to ten observations per freely estimated parameter, is
technically satisfied here (N=726 against a rough count of around sixty parameters for a model this size
gives a ratio near 12), so feasibility is not the reason. The reason is that this notebook's combination
of Section 9's factor-analytic measurement model, Section 12's multivariate structural regression, and
Section 24.1's residual-correlation check already estimates the same two conceptual pieces a SEM would,
a measurement model relating items to latent factors, and a structural model relating those factors to
outcomes, without the added estimation complexity, convergence sensitivity, and more limited Python
tooling (SEM is considerably more mature in R's `lavaan` than in any current Python package) that a single
combined SEM would introduce. Decomposing the problem into these two already-validated pieces is treated
here as the more transparent and reproducible choice for this specific dataset, not as a claim that SEM
would be inappropriate in general.

</div>

<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">24.3</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">A Bayesian Re-Analysis of the Primary Finding</span>
</div>


**Objective.** Re-estimate the rock/metal-depression coefficient in a Bayesian framework and report a
credible interval alongside Section 17's bootstrap confidence interval, making the conceptual difference
between the two concrete rather than asserting it abstractly. **Method.** Conjugate Bayesian linear
regression (Section 3.11): a weakly informative Normal prior on each coefficient (variance 10, wide
relative to the outcome's own scale) and an Inverse-Gamma prior on the residual variance, which together
admit a closed-form posterior, no MCMC sampling required.


```python
prior_variance = 10.0
a0, b0 = 2.0, 2.0

bayes_data = df[['Depression'] + full_spec].dropna()
y_bayes = bayes_data['Depression'].values
X_bayes = sm.add_constant(bayes_data[full_spec]).values
n_bayes, k_bayes = X_bayes.shape

XtX = X_bayes.T @ X_bayes
prior_precision = np.eye(k_bayes) / prior_variance
posterior_cov_unscaled = np.linalg.inv(XtX + prior_precision)
posterior_mean_vec = posterior_cov_unscaled @ (X_bayes.T @ y_bayes)

a_n = a0 + n_bayes / 2
b_n = b0 + 0.5 * (y_bayes @ y_bayes - posterior_mean_vec @ (XtX + prior_precision) @ posterior_mean_vec)
sigma2_post_mean = b_n / (a_n - 1)
posterior_cov = sigma2_post_mean * posterior_cov_unscaled

mf2_col = full_spec.index('MF2_rockmetal') + 1
post_mean = posterior_mean_vec[mf2_col]
post_sd = np.sqrt(posterior_cov[mf2_col, mf2_col])
credible_interval = stats.norm.ppf([0.025, 0.975], post_mean, post_sd)
p_direction = 1 - stats.norm.cdf(0, post_mean, post_sd)

freq_model = sm.OLS(y_bayes, X_bayes).fit()
freq_ci = freq_model.conf_int()[mf2_col]

print(f"Frequentist point estimate:      {freq_model.params[mf2_col]:.4f}")
print(f"Frequentist 95% CI:               [{freq_ci[0]:.4f}, {freq_ci[1]:.4f}]")
print(f"Bayesian posterior mean:          {post_mean:.4f}")
print(f"Bayesian 95% credible interval:   [{credible_interval[0]:.4f}, {credible_interval[1]:.4f}]")
print(f"Posterior probability of direction (coefficient > 0): {p_direction:.5f}")
```

<div style="background:#f5f3ff;border-radius:12px;padding:20px 26px;border-left:5px solid #7c3aed;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;margin:14px 0;">
<h4 style="color:#5b21a6;margin:0 0 10px;">Results: the two frameworks agree closely here, for an instructive reason</h4>

The Bayesian posterior mean (0.504) and credible interval ([0.320, 0.688]) sit almost exactly on top of
the frequentist estimate and confidence interval, and the posterior probability that the true coefficient
is positive is, to five decimal places, 1.00000. This close agreement is itself the expected and
informative result, not a coincidence to note in passing: with N=726 and a weakly informative prior, the
likelihood dominates the prior by a wide margin, which is exactly the regime in which Bayesian and
frequentist point estimates are expected to converge. The two intervals remain conceptually distinct even
where they are numerically similar, precisely the distinction Section 3.8 draws: the frequentist interval
is a statement about the procedure's long-run coverage across hypothetical repeated samples, while the
credible interval is a direct posterior probability statement about this coefficient given the data
actually observed, and the posterior-probability-of-direction figure above is a quantity a p-value simply
does not provide, a direct, easily communicated statement that the evidence for a positive association is
about as strong as this kind of evidence gets.

<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Causal Reasoning and a Directed Acyclic Graph</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">Being explicit about what this design can and cannot identify</p>
</div>


Every association reported in this notebook, however statistically robust, is consistent with at least
three distinct causal stories, and the cross-sectional, observational, self-reported design used here
cannot distinguish between them. Naming this plainly is not a hedge added after the fact, it follows
directly from H1's own wording in Section 2.2, which was phrased around "association" throughout
specifically because a causal claim was never something this design could support.


```python
fig, ax = plt.subplots(figsize=(10, 6.5))
ax.set_xlim(0, 10); ax.set_ylim(0, 7); ax.axis('off')

nodes = {
    'Age': (1.5, 5.5), 'Behaviour': (1.5, 1.5),
    'Music\npreference': (5, 5.5), 'Mental\nhealth score': (8.5, 3.5),
    'Unobserved\nfactors': (5, 0.8),
}
for name, (x, y) in nodes.items():
    color = '#ede4fb' if name != 'Unobserved\nfactors' else '#fdf4ff'
    edgecolor = '#7c3aed' if name != 'Unobserved\nfactors' else '#a21caf'
    style = 'round,pad=0.5' 
    box = plt.matplotlib.patches.FancyBboxPatch((x-0.9, y-0.5), 1.8, 1.0, boxstyle=style,
                                                  facecolor=color, edgecolor=edgecolor, linewidth=1.8)
    ax.add_patch(box)
    ax.text(x, y, name, ha='center', va='center', fontsize=10.5, color='#3b0764', fontweight='bold')

def arrow(start, end, style='-|>', color='#4c1d95', linestyle='solid'):
    ax.annotate('', xy=nodes[end], xytext=nodes[start],
                arrowprops=dict(arrowstyle=style, color=color, lw=1.8, linestyle=linestyle,
                                 shrinkA=35, shrinkB=35))

arrow('Age', 'Music\npreference')
arrow('Age', 'Mental\nhealth score')
arrow('Behaviour', 'Music\npreference')
arrow('Behaviour', 'Mental\nhealth score')
arrow('Music\npreference', 'Mental\nhealth score')
arrow('Unobserved\nfactors', 'Music\npreference', linestyle='dashed', color='#a21caf')
arrow('Unobserved\nfactors', 'Mental\nhealth score', linestyle='dashed', color='#a21caf')

ax.text(5, 6.6, 'Figure 27. A directed acyclic graph for the music-mental health relationship',
        ha='center', fontsize=13, color='#5b21a6', fontweight='bold')
ax.text(5, -0.15, 'Solid arrows: measured in this survey.  Dashed arrows: unmeasured, e.g. personality, life stress, genetics.',
        ha='center', fontsize=9, color='#57534e', style='italic')
plt.tight_layout(); plt.show()
```

<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">26.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Reading the DAG: Three Explanations This Design Cannot Separate</span>
</div>


**Music affects mental health.** Listening to a particular kind of music genuinely shapes mood or
symptoms over time, the causal direction most public discussion of this survey assumes by default.

**Mental health affects music preference (reverse causality).** Someone already experiencing depressive
symptoms may gravitate toward certain genres, lyrically darker or more intense music resonating with an
existing mood, rather than the music producing the mood.

**An unmeasured common cause drives both.** Age partially fills this role and is measured and adjusted
for throughout, but personality traits, life stress, socioeconomic circumstances, or genetic
predisposition to both a temperament and a musical taste, none directly measured here, could each
independently produce the rock/metal-depression association this notebook has documented without either
variable causing the other.

A single cross-sectional snapshot cannot rule out any of these three, or some combination of them, and
this notebook does not attempt to. Reporting an association as robust, as Section 18 does for the
rock/metal-depression relationship, is a claim about the strength and stability of the *statistical*
pattern, not a resolved answer to which of these three stories, or what mixture of them, produced it.

<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Psychological and Ethical Considerations</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">Treating a self-report survey score as exactly that, and nothing more</p>
</div>

<div style="background:#fffbeb;border-radius:14px;padding:22px 28px;border:1.5px solid #fde68a;font-family:Georgia,serif;line-height:1.8;color:#78350f;margin:14px 0;">
<h4 style="color:#b45309;margin:0 0 10px;font-size:1.0rem;">What these four numbers are, and are not</h4>

Anxiety, Depression, Insomnia, and OCD in this dataset are single-item, 0-10 self-report ratings collected
in an anonymous online survey, not scores from a validated clinical instrument such as the GAD-7 or
PHQ-9, and not a diagnosis made by any clinician. No result in this notebook diagnoses any respondent, and
none should be read as evidence about what any individual respondent should do, including what they
should listen to. The original notebook's suggestion of a music-recommendation system built on these
associations, and its claim that people with OCD prefer repetitive music because they seek structured
patterns, are both removed from this analysis: the first treats a correlational finding as though it
licensed an intervention, and the second states a specific psychological mechanism the data cannot
support and that was not sourced to any peer-reviewed literature in the original notebook.

</div>

<div style="background:linear-gradient(135deg,#8b5cf6 0%,#5b21a6 100%);border-radius:8px;padding:11px 20px;margin:24px 0 14px 0;">
<span style="color:#ede4fb;font-family:Georgia,serif;font-size:12px;font-weight:700;letter-spacing:1px;">27.1</span>
<span style="color:#ffffff;font-family:Georgia,serif;font-size:17px;font-weight:600;margin-left:12px;">Threats to Validity, By Name</span>
</div>


**Self-report and recall bias.** Every variable in this dataset, including daily listening hours and BPM,
is what a respondent remembers or estimates about themselves, not an independently observed measurement,
and both are known to be imperfectly recalled, in either direction.

**Common-method bias.** Every variable was collected in the same survey instrument, at the same sitting,
using the same response format, which can inflate correlations between variables for reasons having
nothing to do with any real relationship between them, purely an artefact of shared measurement context.
This is precisely why Section 7.4 declined to use the survey's own "Music effects" item as evidence about
whether music actually helps, that item and the outcome scores were collected in the same breath.

**Selection and representativeness.** Respondents to an online survey about music and mental health are
self-selected in ways a general population sample is not, plausibly skewed toward people with strong
opinions about music, or currently experiencing symptoms salient enough to answer these questions.
Section 6's age distribution, concentrated in the late teens and twenties, is consistent with this and
limits how far any finding generalises to older populations.

**Construct and measurement validity.** A single 0-10 item is a coarser, less validated measurement of
anxiety or depression than a multi-item clinical scale, and Section 13's finding, that treating these
scores as ordinal rather than continuous changes little, speaks to internal consistency across two
statistical treatments, not to whether the item itself captures the intended clinical construct.

<div style="background:#fdf4ff;border-radius:14px;padding:22px 28px;border:1.5px solid #f0abfc;font-family:Georgia,serif;line-height:1.8;color:#701a75;margin:14px 0;">
<h4 style="color:#a21caf;margin:0 0 10px;font-size:1.0rem;">Design and measurement</h4>

This is a cross-sectional, self-reported, observational survey; Section 26's DAG applies to every
association in this notebook without exception. The four outcomes are single-item scores, not validated
multi-item clinical instruments, and several genres (Classical, Pop, Country) showed low communality in
Section 9, meaning the four-factor representation captures their relationship to mental health less
completely than it does for Rock, Metal, or the eclectic-listening genres. The sample skews young
(Section 6), and every finding's generalisability to older populations is correspondingly limited.

</div>

<div style="background:#fdf4ff;border-radius:14px;padding:22px 28px;border:1.5px solid #f0abfc;font-family:Georgia,serif;line-height:1.8;color:#701a75;margin:14px 0;">
<h4 style="color:#a21caf;margin:0 0 10px;font-size:1.0rem;">Statistical scope</h4>

Sixty-four genre-by-outcome tests were run and corrected for in Section 16; the interaction tests in
Section 15 and the GAM in Section 14 were pre-specified but still add to the effective number of looks
taken at this data, and a fully pre-registered analysis would have committed to every specification in
Section 18's robustness table before seeing any result, which this notebook, built and iterated on the
same data it analyses, cannot claim to have done. The proposed framework in Section 24 approximates a
joint multivariate GAM by fitting four separate univariate GAMs and comparing residuals afterward, a
reasonable approximation given Python's current multivariate-GAM tooling, but not identical to a single
jointly-estimated likelihood.

<div style="color:rgba(255,255,255,0.55);font-size:11px;letter-spacing:4px;text-transform:uppercase;margin-bottom:8px;">Section 29</div>
<h2 style="color:#ffffff;font-family:Georgia,serif;font-size:24px;margin:0;font-weight:700;">Conclusion</h2>
<p style="color:#e4d4fb;margin:6px 0 0 0;font-size:13.5px;">What this analysis supports, suggests, leaves uncertain, and cannot establish</p>
</div>

<div style="background:linear-gradient(135deg,#faf8fd 0%,#f3edfa 50%,#faf8fd 100%);border-radius:16px;padding:34px 42px;font-family:Georgia,serif;line-height:1.85;color:#2e1a47;border:1.5px solid #ddc9f2;margin:20px 0;">

<p style="margin:0 0 16px 0;"><strong>What the data support.</strong> After adjustment for age and behavioural covariates, a latent
rock/metal-listening factor is positively and significantly associated with self-reported depression, an
association that survived every check this notebook applied: five alternative model specifications
(Section 18), removal of the most influential observations and a robust M-estimator (Section 19),
continuous and ordinal outcome treatment (Sections 11 and 13), a joint multivariate test (Section 12), a
false-discovery-rate correction across all sixty-four genre-level comparisons (Section 16, where the raw
Rock-Depression and Metal-Depression correlations both survived independently), and a Bayesian
re-analysis whose posterior probability of a positive effect rounded to 1.00000 (Section 24.3). Its
magnitude is real but modest, a standardized coefficient of 0.207, roughly midway between Cohen's
conventional small and medium benchmarks, and a bootstrap 95% confidence interval of [0.318, 0.678] on
the raw depression scale (Section 17). The same factor is significantly associated with anxiety and
insomnia in the less-adjusted specifications, though these associations were not the primary focus of the
robustness battery applied to the depression finding specifically.</p>

<p style="margin:0 0 16px 0;"><strong>What the data suggest, with more caution.</strong> An electronic/digital-listening factor showed a significant
association with depression before adjustment that attenuated to non-significance once behavioural
covariates were added (Section 11), while the same factor's association with insomnia did not attenuate,
an outcome-specific pattern rather than a uniform one. Regularized regression on the raw genre variables
(Section 20) additionally surfaced a negative association between Country listening and depression that
had not appeared in the univariate FDR-corrected tests, a genuinely new signal visible only once every
genre competed for predictive credit simultaneously, but one that has not yet been checked against the
same robustness battery the rock/metal finding passed, and is reported here as a lead rather than a
confirmed finding. A relationship between age and depression is real but nonlinear (Section 14), rising
through the twenties and declining through middle age before an upturn at the oldest ages that rests on a
small number of respondents and should be read cautiously.</p>

<p style="margin:0 0 16px 0;"><strong>What remains uncertain.</strong> Music-preference information improves cross-validated predictive performance
for a high-depression classification only modestly, from a Random Forest ROC-AUC of 0.614 without it to
0.635 with it (Section 21), and a neural network did not outperform simpler models at this sample size
(Section 22), consistent with a real but modestly-sized effect rather than a complex, learnable nonlinear
signature. Whether a semantic representation of genre labels drawn from outside this survey would add
further information could not be verified inside this notebook's own execution environment (Section 23),
and remains an open question for whoever runs this analysis with internet access enabled.</p>

<p style="margin:0;"><strong>What the data cannot establish.</strong> Nothing in this notebook distinguishes whether music preference
influences mental health, mental health influences music preference, or both reflect some third,
unmeasured factor (Section 26); the cross-sectional, self-reported, single-timepoint design was never
capable of adjudicating between these three explanations, and no amount of additional statistical
machinery applied to the same snapshot changes that. Separately, and independently confirmed in Section
24.1's proposed framework, the correlation among the four mental-health outcomes themselves is almost
entirely unexplained by music, age, or behaviour, whatever produces that shared variance lies outside
this survey's measured variables entirely. No result here diagnoses any individual, recommends any
listening behaviour, or should be read as a reason to change what anyone listens to.</p>
